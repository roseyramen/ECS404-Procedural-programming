## Question 1 \[10 marks\]

**Explain** what is meant in programming by:  

i) a base case  

ii) a step case

**Illustrate** your answer using the example method sum given below.

```java
public static void sum(int n)
{
   if (n == 0)
   {
           return 0;
   }
   else
   {
        return (n + sum(n-1)));
   }
}
```
As part of your answer **explain** whether or not it is possible to have multiple base cases or multiple  step cases.

### Example answer

These terms relate to recursion, where a method calls itself as a way to do repetition. The base case is the terminating case of the recursion. The step case is the recursive case: the case that leads to the state moving closer to the base case.  

Considering the simple recursive method sum. It adds the first `n` numbers:
```java
public static void sum(int n)
{
   if (n == 0)
   {
           return 0;
   }
   else
   {
        return (n + sum(n-1)));
   }
}
```
Here the base case is that `n` is `0`. In this situation the code returns an answer directly. Method `sum` does not need to do any further recursive calls (ie calls to itself) in this case.

The base case acts like the terminating condition of a while loop (or more specifically its negation given a while loop condition states when to continue rather than when to stop). Without a base case, a recursive method would never stop. There can be multiple base cases in a recursive method. FOr example in the above example rather than recurse we could have another base case of when `n` is 1 as the answer is trivial then to, needing no calculation giving:

```java
public static void sum(int n)
{
   if (n == 0)
   {
           return 0;
   }
   else if (n == 1)
   {
           return 1;
   }
   else
   {
        return (n + sum(n-1)));
   }
}
```
This now has two base cases: the cases when `n` is 0 and when `n` is 1

The step case is the case (in the original example) when `n` is not `0`. In the step case a recursive call is made with new arguments taking the method a step closer to solving the problem (here `sum(n-1)`). It takes a step closer to the base case in the sense that the recursive call is now with the value `n - 1` and that is a recursive step closer to the base case of `n` being `0`. By taking a series of step cases that always move towards the base case, you should be guaranteed to eventually hit the base case allowing the method to terminate. There can be multiple step cases in a method too. For example in a binary search method one step case deals with the key being less than the mid point and one deals with it being greater than the mid point.

## Question 2 \[10 marks\]

i) **Write** an equivalent method to the following that does not use a recursive call.
```java
public static int f(int n)
{
   if (n == 0)
   {
       return 1;
   }
   else
   {
       return (n * f(n-2));
   }
}
```

ii) **Justify** that your code is equivalent when the method is given a positive even number as argument.

iii) **Explain** how the above recursive program could fail to terminate.

### Example answer


````java
// i)
public static int f1(int n)
{
    int total = 1;

    while (n != 0)
    {
        total = n * total;
        n = n - 2;
    }
    return total;
}
````

ii)

The original code multiplies every second number working down from that given, down to 1. 
This code can be seen to be equivalent by considering cases. 

The simplest case is the base case when n is 0. The recursive code returns 1 in this case. In the new version the total is set to 1 and when n is 0 the test is immediately false so does not execute the body. 

Next consider the case where n is an even number (2, 4, 6, ...) In any of these cases the recursive version multiplies that number by the recursive result so by the next smallest even number and all those less than it until 0 is reached when it is multiplied by 1 instead eg

f(8) = 8 * f(6) = 8 * 6 * f(4) = 8 * 6 * 4 * f(2) = 8 * 6 * 4 * 2 * f(0) = 8 * 6 * 4 * 2 * 1

For any even number is multiplies together all the smaller even numbers together.
The while loop version does the same as it starts with the total at 1 and multiplies it by n before subtracting 2 and continuing until n becomes 0. This means the same calculation is performed
f1(8) = 1 * 8 * 6 * 4 * 2. The loop terminates when n is 0 so it is not multiplied and the result is the same.




iii)

It could fail to terminate if given an odd number as argument. The only base case is when `n` is `0`, but if `n` starts as an odd number then it will never hit exactly `0`, because the step case subtracts `2` each time. So once it gets to `1`, the step case will call `f(-1)`  which will call `f(-3)` and so on moving away from the base case now rather than towards it. A new base case is needed to cover `n` being a negative number.

## Question 3 [10 marks]

**Discuss** the truth or otherwise of the statement "Recursion is not necessary".

### Feedback on the kind of answer expected

*A question like this asking for a discussion of the truth or otherwise is asking for an essay-style answer that has a clear introduction, body and conclusion and that looks at arguments for both sides before drawing a conclusion.*

### Example answer

Recursion involves a method calling itself so as to provide a form of repetition in a program, executing some sequence of instructions repeatedly until some terminating condition is met. In this way it is equivalent logically to a while loop.

Recursion and while loops therefore do the same job in a programming language. If a language has a general while loop construct then there is no absolute need for any other kind of loop construct, including recursion. Anything you want to program, where you might consider using recursion, can be programmed using the while loop you have. In general the while loop version would also likely be more efficient as it avoids the overhead of method calls.

However, there are advantages of recursion, so much so that some whole languages (functional programming languages) make do without loop constructs at all and just use recursion alone. In particular, recursion is for many problems a very natural way to think about solutions and so to write programs.  For example, a mathematician would naturally define factorial numbers recursively:

1! = 1  
n! = n  x (n-1)!

Given a specification like this it is much easier to write a method to compute factorial recursively rather than having to rethink it in terms of a loop. The code is almost identical up to syntax to the way you have written the specification. Divide and conquer algorithms such as binary search and mergesort are a good example of this. The natural way to explain them is in terms of recursion: to do a binary search of an array go to the middle and check if it is larger or smaller than the key. If it is larger then binary search the lower half of the array, and otherwise binary search the upper part. This naturally translates directly in to a recursive method.

If the natural way to think of a problem is recursively then it makes sense to code it recursively. It is much quicker to do. You are also less likely to make mistakes given you do not need to make a large transformation of the problem to get to the solution. This style of thinking about problems is also one that help avoid bugs in general, as once you are used to the style of thinking it can be easier to get things right.

Thus recursion is not necessary in that if a language does not have it, then there are no programs that become impossible to write. However it is very useful and certainly can help in helping to write bug free code more easily.
