## Introduction
We have seen how **while** loops can control repetition in a variety of ways where the number of times the body of the loop is executed (the number of **iterations**) is unpredictable including with various complex tests.

We have also seen how a while loop can actually do anything a for loop can do. In particular, you can write a counter- controlled loop using a while loop. However, counter-controlled loops are better written using a for loop as it immediately indicates to a reader that it *is* a counter-controlled loop so is easier to understand. Also, by pulling the three key parts of a counter-controlled loop into one line (initialising the counter, testing the counter and changing the counter), means they are less likely to be forgotten.

Now, we look at how we can have a special value in the input stream to act as a terminating value saying there is no more data.

## Avoiding doing the same exit test twice

The solution we have given to do this works but is inelegant. It does the same test twice, once in the while loop and once in the extra if statement. That is a waste of time and makes the code more complicated than needed so harder to read. Worse it would mean we need to modify the exit test in two places if we changed how the loop ended. That could lead to an inconsistency and subtle bug if changes were needed later.

There is a better way - we "unroll" the loop. This just means we pull the first step out of the loop body. We
- Do the first input OUTSIDE the loop.
- Make the LAST step of the body of the loop to input the next value

This means that the same sequence of input and processing steps are done just as before.

## A general pattern for sentinel based input

The general pattern we are using here is

```
    <do first input>
    
    while (<check for sentinel value to terminate>)
    {
         <process the value just input>

         <do next input as last step of loop>
    }
```

This is a very useful pattern to use whenever a termination symbol is part of the stream of values being input and then processed but where you do not want that value to be processed.

## A sentinel value of a different type...

The above is a big improvement but what if we do not want the person to type a numeric code to mean quit. What if we just want the user to type quit instead of a number.

At first sight that seems to be a problem. Doesn't the strict type system of Java disallow that. quit is not a number! Well no. When we input values they all start off as Strings. We just have to make sure that we do not treat it as a number until we know that they did not type in the special quit value. 

The principle is then no different to what we are already doing. It is just that converting the value to an integer becomes part of the processing phase rather than part of the input phase. 

This is a reason why it is good to input things as Strings and not use the less flexible library methods. The solution we get is still elegant.

## Other Data streams

The series of values that contain the sentinel value do not have to come from input. They could, for example, come from dice rolls, be characters in a string of text, or be already stored in an array (or as we will see later be stored in a file).

The principle, however, is the same and you can use essentially the same pattern of while loop. You get or generate the first value outside the loop, process the most recent value at the start of the loop body and then get or generate the next value as the last step of the loop.

### Solution to Exercise 4<a id="6-3-Solution4"></a>


The unrolled loop works because the same instructions are still executed in the same order. First we input an int then add it to the score, then input an int then add it to the score ... It is just that the position of the loop test in this sequence has been moved to be immediately after doing the input every time (either immediately after the first input to decide to enter the loop body at all, or immediately after the input at the end of the body as the test is always done straight after the body is executed. Because we do the loop test in the right place in the sequence now, we don't need the extra if statement to do it.

This means that we check the input to see if it tells us to terminate as the next step after it is input and before we do anything with the value input. If it does mean "end the loop" then we end the loop immediately. If not we enter the body and process the value just input, which we now know is an actual value needing processing.

### Solution to Exercise 6<a id="6-3-Solution6"></a>


The method first inputs a String (NOT an integer!) using inputString and then immediately does the loop test. If what is typed was the String, QUIT, then it exits the loop immediately so never does the body, so never tries to convert it to an integer. It avoids crashing from a type error as it never tries to convert "QUIT" into an int.

On the other hand if QUIT was not typed it enters the loop body. The text input was not "QUIT" so we can take it that it was a number (we are assuming only numbers or QUIT are entered). It is therefore safe at this point to convert the String to a number (really we should check at this point it is a number and not something else but we will cover that input validation in a later notebook).

Once converted to a number we just process it as before. Then as the very last step of the loop we input another String  (again  at this point not an integer). As the next thing we do after reaching the end of the body is the loop test, if it is QUIT then we will finish immediately and avoid crashing: we will not try to convert the String "QUIT" to a number. Otherwise, we enter the body again, convert it to a number and process it...and so on.

In this way we have ensured we never try and turn the String "QUIT" into an integer, but when numbers are input we continue as normal processing the number input - so adding it to the total.

### Solution to Exercise 12<a id="6-3-Solution12"></a>


A While loop can be used to process a stream of data where one or more special values in that stream mean to stop processing any more data. The stream of data could be values input by the user, stored in an array, come from a file, or be generated by some function such as being randomly generated. 

The processing part could be doing anything such as adding numbers, filtering values to print, or doing some other kind of accumulation. 

The special value that is used to terminate the input could be a value of the same type as the rest of the value (eg a number in a stream of numbers). In this case, it must be a value that cannot be a valid piece of data (eg -1 if the data represents ages so positive numbers, or 999 if the numbers are the age of humans). Otherwise it would not be clear whether it was an actual value or the special value meaning to end. This special value is called a sentinel value (as it stands out from the others).

When we get to that sentinel value in the input stream we stop the loop and do not process it. The cleanest way to do this is to "unroll" the loop  following the general pattern:

```
data = <get first value>
while( data is NOT the sentinel)
{
   <process data>

   data = <get next value>
}
```

Here, the first value in the stream is input outside the loop. We then immediately check if it is the sentinel as we enter the loop as the test of a while loop is always done first. If it is the sentinel then the loop body is not executed. This covers the case when there is no data in the input stream, just the terminating sentinel value. This might happen for example when processing the characters of an empty String one at a time. A string ends when the terminating `\0` character is encountered, and for an empty string it is the only character.

In the body of the loop,  the current value is processed next so initially that first value. Once that is done, as the last step of the body it then fetches the next value in the data stream. Again, this means that the next thing done is to check if it is the sentinel value as the loop jumps back to the test at the end of the body.

This ensures that after EVERY value fetched from the data stream, we immediately check if it tells us to terminate, and do so if it is. That ensures we do not erroneously process that sentinel value which we don't want to do as it is not actual data (eg if -1 is used in a stream of ages we are finding the average of, then we do not want to add -1 into the average!)

Below is a concrete example. If we are inputting values with sentinel -999 and just printing the data value as the processing step, then filling in these concrete steps into the pattern it becomes:

```java
int data = inputInt("First value:");

while(data != -999)
{
   print(data);

   data = inputInt("Next value:");
}
```

By unrolling the loop in this way so that the first input is outside the loop we avoid  complex code with more complicated structures including an embedded if statement doing the same test as the loop itself.

Good decomposition can mean writing a special method to check for the terminating sentinel value(s). This can be especially useful if there is more than one or it may change, to abstract away from the details.

The sentinel value itself should generally be made a final variable to give it clear meaning eg

```java
final int QUIT = -1;
...
while(data != QUIT) ...
```
In some situations there is no possible sentinel value of the same type as the data - e.g. the data values could be any integer so no integer can be a sentinal calue. Also using termination codes does not make for a good user interface. They are harder to remember. It can be easier for a user to remember the terminating value if they have to type is something like END or QUIT.

The senitinel value being a different type does not change the general pattern. It just means that the data stream is treated as a stream of Strings. The conversion from a String to another type (eg converting the String to an int) is not done until the processing step rather than being done as part of the input step. For example, we might have 

```java
int total = 1;
final String QUIT = "END;
String data = inputString("First value:");

while(!data.equals(QUIT))
{
   int data_as_int = Integer.parseInt(data)
   total = total * data_as_int;

   data = inputString("Next value:");
}
```

Here we input the data as a String and only inside the body at a point where we know it is not the terminating value do we convert it to a number (using Integer.parseInt) so we can do arithmetic on it (here multiplying it by the previous numbers). By treating the conversion as part of the processing step we can use exactly the same while loop pattern, just using String equality.

Overall then, sentinel controlled loops provide a way to process data with the indication of when to stop in the data itself. This can be done using a simple while loop structure as long as the first input is done outside the loop.
