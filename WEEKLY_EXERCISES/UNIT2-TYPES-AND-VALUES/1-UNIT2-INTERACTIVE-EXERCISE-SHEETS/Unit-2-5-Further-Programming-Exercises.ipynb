We have added to this a commented out line that shows a new way to print that formats numbers
controlling the number of decimal places printed.

```
System.out.format("Each of the %.0f employees gets %.2f pounds each.", 
                  EMPLOYEES, profit_each);   
```
The method ```System.out.format``` takes a String argument that now has formatting instructions in it that control where to print the value of variables and how many decimal places are printed.
The positions of variables are now indicated by a code starting with a %. %.0f means print the decimal number with no decimal places printed. %.2f means print the decimal number with two decimal places printed. The variables whose values are to be printed appear after the String and are used in  the positions of the String in order.
