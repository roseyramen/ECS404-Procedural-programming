## Introduction
Good style is about writing code in a way that makes it easier to read and so understand and so both get right originally as well as modify without introducing mistakes later.

Here we look at how variables contribute to readability. 

There are lots of aspects to good style but the basics are:
- write useful comments, not too many, not too few
- indent your code and line up curly brackets
- use blank lines to make separate blocks or methods stand out as separate
- choose good, meaningful variable names
- choose good meaningful method names
- use final variables for literal values

More sophisticated forms of good style that are critical to making code readable are good: decomposition (breaking a program into methods corresponding to sub-tasks), generalisation (writing general methods that can be used for several related tasks) and abstraction (hiding the detail of things so you can focus on what is important with the coding task at hand). The use of these form the foundation of what is meant by computational thinking.

## Naming literal values

*Literal* values are just actual values that appear in your code  like 100 or 12000 or 0.5. It is often a good idea to replace literal values with variables, particularly when it is a number that means something. This gives that number a name and if that name is well chosen can make clear the value's meaning.

Here is an example from school. Suppose you want to work out the perimeter of a regular Octagon. The building called The Octagon at QMUL is octagonal in shape, for example. The perimeter is just how far you would have to walk if you were to walk round its edges.  As all the sides are the same (that's what it means for it to be regular) and an octagon has 8 sides, the perimeter is just 8 times the side length.

The following method calculates this for us for any side length we give it. Read it and make sure you understand it.

````java
// Calculate the perimeter of a regular octagon
//
public static int perimeterOfOctagon1 (int side_length)
{
    return 8 * side_length;
}
````

````java
// Test above method
System.out.println( perimeterOfOctagon1 (1) );
System.out.println( perimeterOfOctagon1 (3) );
````

However, as in our description that number 8 in the method has a meaning - it is the number of sides. That is why we are multiplying by 8 rather than 6 or 21. So it makes clearer what the method is calculating if we actually call it by a meaningful name e.g. OCTAGON_NUMBER_OF_SIDES, as in the following better version of the same method.

````java
// Calculate the perimeter of a regular octagon
//
public static int perimeterOfOctagon2 (int side_length)
{
    int OCTAGON_NUMBER_OF_SIDES = 8;
    
    return OCTAGON_NUMBER_OF_SIDES * side_length;
}
````

````java
// Test above method
System.out.println( perimeterOfOctagon2 (1) );
System.out.println( perimeterOfOctagon2 (3) );
````

Now we have made it absolutely clear that that number is meant to be the number of sides of an octagon. 

It is a good idea to write variables that represent unchanging values like this in all capitals as we have done here. That gives you a visual reminder when reading the program that they do not change.

## Final Variables ##

We can do better than this though, by using final variables. Final variables are just variables that have been declared to never change. They represent constant values. They can be initialised but never then changed. No assignment is allowed after the first one. Any variable that is being used to represent a constant literal value like this should be declared final. The program works exactly the same. It just makes it clear that that variable is never intended to change.

You do this by simply putting the keyword *final* in front of the declaration, so if you were declaring a variable representing the teams in a knockout competition to be 64 as

int NUMBER_OF_TEAMS = 64;

then you would instead write

final int NUMBER_OF_TEAMS = 64;

A better version of our method to calculate the perimeter of an octagon would use a final variable like this. Notice the use of the keyword final in the method below.

````java
// Calculate the perimeter of a regular octagon
//
public static int perimeterOfOctagon3 (int side_length)
{
    final int OCTAGON_NUMBER_OF_SIDES = 8;
    
    return OCTAGON_NUMBER_OF_SIDES * side_length;
}
````

````java
// Test above method
System.out.println( perimeterOfOctagon3  (1) ); // should give answer 8
System.out.println( perimeterOfOctagon3  (3) ); // should give answer 24
````

## Why use final variables ##

### Improving readability, Spotting mistakes ###
Final variables improve readability because they make clear the meaning of numbers in each context they are used. By using final variables, we are also giving information to the compiler that can help it spot mistakes we make. The compiler has been told that the final variable should not change. Therefore if you write an assignment that tries to change it, then the compiler can warn you that this is a mistake.  This makes the mistake an easy to spot compile-time error instead of it being a hard to spot run time error. Run-time errors may never be detected and in end up in production programs. With a compile time error, you do not have a program at all to run as the compiler refuses to compile it.

### Making changes ### 

Another reason for using final variables is that it makes it easier to make changes without introducing mistakes. This might be to make a constant like pi and so our program more accurate, or just because some requirement of the program has changed. 

For example, suppose you are writing code to help control a chocolate factory production line. Small bars are made of 8 triangles of chocolate and big bars are made of 16 triangles of chocolate with the triangles themselves the same size: each triangle is made of 12g of chocolate  We might write a method like this to do the calculations for us.

````java
// Work out how much chocolate is needed to make small and big bars of chocolate
// Each triangle uses 12g
//
public static void amountOfChocolate1()
{
    int small = 12 * 8;
    int big = 12 * 16;
    
    System.out.println("A small bar needs " + small + "g of chocolate.");
    System.out.println("A big bar needs " + big + "g of chocolate.");
    
    return;
}
````

````java
// Test the above method
amountOfChocolate1();
````

As we have already seen having all these numbers makes the code hard to read, hard to understand. It would be easy to get them confused. If we needed to reduce the number of triangles in a big bar (to make them cheaper) we could easily change the wrong number. Similarly, it is better to avoid making changes in multiple places.
