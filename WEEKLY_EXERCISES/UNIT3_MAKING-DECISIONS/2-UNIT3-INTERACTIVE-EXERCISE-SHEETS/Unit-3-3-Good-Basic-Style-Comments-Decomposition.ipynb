## Introduction
Good style is about writing code in a way that makes it easier to read and so understand and so both get right originally as well as modify without introducing mistakes later.

There are lots of aspects to good style but the basics are
- write useful comments, not too many, not too few
- indent your code and line up curly brackets
- use blank lines to make separate blocks or methods stand out as separate
- choose good, meaningful variable names
- choose good meaningful method names
- use final variables for literal values

More sophisticated forms of good style that are critical to making code readable are good: decomposition (breaking a program into methods corresponding to sub-tasks), generalisation (writing general methods that can be used for several related tasks) and abstraction (hiding the detail of things so you can focus on what is important with the coding task at hand). The use of these form the foundation of what is meant by computational thinking.

In this notebook we focus first on good commenting of code which also leads to some simple points about decomposition. We will see how thinking about comments can help improve the decomposition of code into methods.

## Start the program with a comment ##

*All* programs should start with a comment at the very start (outside the class and before any libraries are loaded). These whole program comments include at the very least what the program is for, the name of the author (in tests and exams you can use your Student id), the date it was written (and ideally a version number). These can be written using block comments like this that start with /* and end with */:

````java
/*
 * A program to act as my intelligent digital assistant
 *
 * AUTHOR: Tony Stark
 * DATE: 17 January 2008
 * Version: 1
*/
````

The extra stars are just there to make the comment block stand out.

An alternative is to comment the start of each line (again the stars are just used to make it stand out).

````java
// ****************************************************
// A program to act as my intelligent digital assistant
//
// AUTHOR: Tony Stark
// DATE: 17 January 2008
// Version: 1
// ****************************************************
````

## Do not over comment: the wrong kind of comment ##

A common novice mistake is to over-comment. In the worst case this means writing so many comments that the code itself can no longer be easily seen. Also it often involves writing the wrong kind of comments. Comments should not say what individual lines do but what a logical block of code does. Comments such as the following are useless in real programs. They are often used in teaching to indicate to a novice what a line does. However, real comments you write should be for people who can code, so should not just repeat in English what is easily understandable from just reading that line (if the reader is someone who can program). Comments should give the bigger picture so that they tell you what the code does overall, so by reading one comment you understand what a whole block of code does.

One exception to this is that in the style we follow we comment the final close bracket of methods and the class to say what they end. This is because it helps novices focussing on where methods end, and helps ensure close curly brackets are not missed.

**You should NOT write comments like any of the ones below in your programs (apart from the one noting the end of the method), or comment each line like this.**

````java
public static void printTemperatures () //create printTemperatures method
{
    int temperature = 0; // declares and sets temperature variable to 0
    
    System.out.println("It is cold."); // prints It is cold to the screen
    temperature = 32; // assigns 32 to temperature
    System.out.println("It is now really hot."); // prints It is now really hot.

    return; // return from method
} // END printTemperatures
````

````java
printTemperatures (); // call method printTemperatures
````

**ALL the above comments are bad comments.**

If you were writing comments for code like this it would be better to comment the section of code. If you break your code into small methods then that means just commenting the method, like this:

````java
// Set a cold temperature and print that it is cold
// before changing it to hot and printing a message about that
//
public static void printTemperatures ()
{
    int temperature = 0;
    
    System.out.println("It is cold.");
    temperature = 32;
    System.out.println("It is now really hot.");
    
    return;
} // END printTemperatures
````

````java
printTemperatures ();
````

Notice that no individual lines are commented here. The method as a whole is commented with a comment that gives a bigger picture of what the method does overall.

## Comment tricky bits of code ##

The one situation where you might comment an individual line is if it is doing something a little tricky, so that a good programmer might not immediately see what was happening. One example like this might be when using the modulus (remainder when dividing) operator to do clock arithmetic such as in the following. The idea here is to have a test for even numbers so every second day is identified. Whilst a programmer might understand that % does clock arithmetic they may not immediately see why it is done here without a comment.

````java
public static void tabletDays (int day)
{
    if (day % 2 == 0) // distinguish odd and even days as only take tablets every second day
    {
        System.out.println("Take a tablet today");
    }
    else
    {
        System.out.println("No tablets today");
    }
    
    return;
}  // END tabletDays
````

````java
tabletDays(5);
tabletDays(8);
````

However, in a situation like this there is often actually a better way than commenting the line. That test above is doing a well-defined (if hard to work out) task. Well-defined tasks are better split off as a method. For example, we can have a method that calculates if it is every second day or not as below.

The method itself can then be commented, rather than the line, and its name makes much clearer what it is doing as in the following. By using good variable and method names, the need for commented lines can often be avoided as the name gives the information a comment would.

We still need a comment to explain the tricky bit of code, but now we are back to following the original rule:

**ONLY COMMENT METHODS NOT LINES**

because we have made better use of methods. Below we leave writing the comments as an exercise for you to do.

````java
public static boolean isEverySecondDay (int day)
{
    return (day % 2 == 0);
} // END isEverySecondDay
````

````java
public static void tabletDays (int day)
{
    if (isEverySecondDay(day) )
    {
        System.out.println("Take a tablet today");
    }
    else
    {
        System.out.println("No tablets today");
    }
    
    return;
}  // END tabletDays
````

````java
tabletDays(5);
tabletDays(8);
````

## Comment blocks ##

Rather than commenting lines it is appropriate to comment blocks of code. That is what we are doing by commenting methods as a method contains a block of code. However, blocks are also used in if statements and as we will see later loops. They could be commented as a whole too.

For example, if two if statements follow each other you might comment each saying what they each do overall.

As with the earlier example, though, there is a better way than writing comments in the middle of a method. The fact that we can write a comment for these two parts is a good indication that they do clearly defined tasks so should be separate methods. The length of the method, which is a lot to take in in one go, also implies it would be better decomposed too. A better solution is therefore not one method but splitting it into three methods as follows (one for each if statement and one to call each in turn. 

Again, this does not remove the need for comments but reinstates the rule that comments should be before methods not inside them.

````java
// Find the position of the colour in the rainbow starting with Red as 1.
// Return -1 as a sentinel value to mean "not known".
//
public static int getRainbowPosition (String rainbowColour)
{
    int positionInRainbow;
    
    if (rainbowColour.equals("Red"))
    {
        positionInRainbow = 1;
    }
    else  if (rainbowColour.equals("Orange"))
    {
        positionInRainbow = 2;
    }
    else if (rainbowColour.equals("Yellow"))
    {
        positionInRainbow = 3;
    }
     else if (rainbowColour.equals("Green"))
    {
        positionInRainbow = 4;
    }
     else if (rainbowColour.equals("Blue"))
    {
        positionInRainbow = 5;
    }
     else if (rainbowColour.equals("Indigo"))
    {
        positionInRainbow = 6;
    }
    else if (rainbowColour.equals("Violet"))
    {
        positionInRainbow = 7;
    }
    else
    {
        positionInRainbow = -1;
    }
    
    return positionInRainbow;
} // END getRainbowPosition
````

````java
// Test the above method
System.out.println(getRainbowPosition ("Yellow"));
System.out.println(getRainbowPosition ("Violet"));
System.out.println(getRainbowPosition ("Black"));
````

As we are using two methods, the first has to pass information to the second. It does this by returning the **value** of the position it has calculated back to whatever calls it. Notice here the variable *positionInRainbow* is local to this method. It does not exist in the other methods. The methods below store the same value in their own different variables, named different things. They could be named the same thing but they would still be different variables, different storage spaces, as all methods have their own storage space not shared with others.

Notice we can now test this method separately (as above, printing the values the method returns) to check that it returns correct values, before moving on to the next method. To be sure we should run tests for every branch in the code to be sure (not just the three above). As an extra exercise add more tests to make sure it always works.

````java
// Given a colour and its position in the rainbow (starting with Red as 1)
// Print a message about the colour and the position it is in
// -1 is a sentinel value to mean "not known".
//
public static void printRainbowMessageGivenPosition (String colour, int position)
{            
    if (position == -1)
    {
        System.out.println("Sorry, that is not a traditional colour of the rainbow.");
    }
    else 
    {
        System.out.println(colour + " is in position " + position);
    }
             
    return;
} // END printRainbowMessageGivenPosition
````

To print the message we need to be given both the colour and position to use in the message to print. We create local variables as arguments and these values must be passed to the method when it is called as in our tests below. As this method prints things rather than returns a value, we do not need print statements in the tests here.

````java
// Test this method
printRainbowMessageGivenPosition ("Red", 1);
printRainbowMessageGivenPosition ("Green", 4);
printRainbowMessageGivenPosition ("Mauve", -1);
````

Here we were just testing the method so had to pick correct pairs of numbers and colours ourself. The next method now puts the above two parts together, by calling the first part to find out the correct number and then passing that information to the second (along with the colour) so it can print the right message.

````java
// Print a message about the position a colour appears in the rainbow.
// -1 is a sentinel value to mean "not known".
//
public static void printRainbowMessage (String rainbowColour)
{
    int rainbowPosition;
    
    rainbowPosition = getRainbowPosition (rainbowColour);
    printRainbowMessageGivenPosition (rainbowColour, rainbowPosition);
             
    return;
} // END printRainbowMessage
````

This is the method that replaces the original one. It does exactly the same thing but in two steps (each a method call) - calling the method to work out what the position is, and store that locally. That allows it to pass the value stored there to the second method. It prints the message.

We can now understand each part (each method) separately and the naming of the methods help us understand how this last method achieves it's effect by first getting the position and then printing a message about it.

Thinking about better commenting has led us to a way to write better code, by decomposing it into parts that we comment separately.

````java
// Test the above method
printRainbowMessage ("Indigo");
printRainbowMessage ("White");
````

A comment at the start makes sure everyone knows what the purpose of this program is and so does not get confused and think it does something it is not intended for (as well as knowing what it should be used for). We not only know what the program is for but also expectations about the way it must be used (stating assumptions a program makes is very important at the start).
