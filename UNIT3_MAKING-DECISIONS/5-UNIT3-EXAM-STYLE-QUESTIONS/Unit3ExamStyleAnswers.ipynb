## Question 1 \[10 marks\]

**Explain** what is meant by each of the following **illustrating** your answer with examples in Java.

i) a type  
ii) a boolean expression

### Example answer

i) a type  
Types are used in programming languages to provide information to the compiler and make programs easier to understand. Each variable is declared with a type and different values have different types. They essentially group values into similar things with the similar storage requirements and similar options. When variables are declared, this allows the compiler to allocate space and also to notice errors when an operation that applies to one type is applied to a value or variable of an inconsistent type. Example types are the integers (...```-3```, ```-2```, ```-1```, ```0```, ```1```, ```2```, ```3```...), booleans (```true```, ```false```) and Strings ```"hello"```, ```"paul"```, etc.
Java uses static typing. This means neither values nor variables can change type. A variable is declared with a type and from that point on, while it exists, has that type. Attempting to store something of a different type in it will raise a compiler error. SO for example, we can declare
```java
String name;
```
and form that point on the variable ```name``` holds string values and nothing else. If we later were to write
```java
name = 5;
```
then the compiler will raise an error
Static types like this are a good thing as it means that the compiler can spot more of our mistakes.

ii) a boolean expression  
Expressions are parts of a program that when executed evaluate to give a value. A boolean expression is an expression that has as result value either true or false: the expression evaluates to true or the expression evaluates to false (and nothing else is possible). Examples include equality tests like (```a==0```). Either ```a``` is ```0``` (```true```) or it is not (```false```). The latter is the result for any value other than 0. Another example is a greater than test of two variables. 

The simplest kind of boolean expression is a boolean variable, so if we declare
```java
boolean is_red;
```
then ```is_red``` can be used as a boolean expression as it evaluates to true or false.

More complex expressions can be built up using logical operators such as ```&``` (and) as in (```a==0 & a>b```) combining smaller boolean expressions in to bigger ones. 

Boolean expressions are mainly used in control structures such as if statements to alter which statements are executed based on the results of a boolean test.

## Question 4 \[10 marks\]

**Explain** what is meant by a *record* and by the *fields* of a record **illustrating** your answer with the example of a record cars.

### Example answer

Records are a simple kind of data structure that join together values of potentially different types in to a single composite value. When we define a record we define a new type for those composite values that can thereafter be stored and manipulate as a single value of that more complex type. It is a way of representing lots of data about a single thing. So for example we might define a new record to represent Cars. We would create a new type called Car. Values of that type would be the combined information about the car. So we could store a car as its make, its registration number and the insurance premium to pay (if these were records held by an insurance company say). A single value would then be thought of as a tuplet of values such as ("Mini", "ABC123D", 730.54) representing a single car. This would be a single value of type Car made of 2 Strings and a double bound and stored together. A different car would be represented by a different tuplet of values. Such a record could by stored in a single variable of type Car.

The fields of a record are the separate compartments of the record.  The fields are the separate component parts. For example, if we define a record for cars for an insurance company’s records as above we would have fields of car make (a string), car registration number (a string)
and a price for the premium (a double).
When we define a new record type we define the field names so the above record type would be defined as follows:
```
class Car {
   String carmake;
   String carregistration;
   double premium;
} // END class Car
```
The fields of the record type are ```carmake```, ```carregistration``` and ```premium```. Those field names can then be used to pull out the individual sub-values of the particular car. So if we had stored the details of a Mini car as above in the variable ```myCar``` then ```myCar.carmake``` would return the make value stored in the record ("Mini"), ```myCar.carregistration``` would return the registration value stored in the record ("ABC123D"), and ```myCar.premium``` would return the insurance premium value stored in the record ("730.54"). Values can be stored in the fields of a record in a similar way using the field names combined with the variable names as though the combined name were variable names.

## Question 5 \[10 marks\]

**Explain** why methods are an important form of abstraction that can help if a program is to be changed, **illustrating** your answer with the example of a print method.

### Example answer

Abstraction is concerned with hiding the details of things. In programming various forms of abstraction are used. Methods provide an important form of abstraction in that they hide the details of how the method is implemented. This is replaced by the name of the method in the rest of the program. All someone using the method writing a program needs to know is what the method does not how it does it. This makes coding much easier and is the foundation of decomposition. It makes changing code easier as the people who need to make the change can understand the code much more easily and if you cannot modify code until you understand it. Methods hide the detail of how things are done replacing them with a name about what is done and that means you can understand what a method is for without having to understand how it is done. If I have written a print method then just from its name I can see its purpose is to print messages, without worrying how it does it (when understanding other parts of the code.

This kind of procedural abstraction also means that implementation details in one method can be changed without affecting the rest of the program as long as the interface remains the same (ie the arguments passed and the type of value returned).

To take a very simple example suppose we write a method print
```
public static void print(String message)
{
   System.out.println(message);
}
```
Having written this, in the rest of the program we only need to know the name print and that it prints the given message to use it. We do not need to remember the message. So when adding new code that needs to print consistently with the rest of the program the programmer can safely use it.

If it is decided later to print messages in pop-up windows not in the console, we can change the implementation of print to:
```
public static void print(String message)
{
   JOptionPane.showMessageDialog(null, message);
}
```
As the interface to print is unchanged this change can be made and the rest of the program still works. Everywhere we print things throughout the whole program  changes consistently just by us changing this one line as long as we have consistently used the print method.

## Question 6 \[10 marks\]

**Compare and contrast** procedural programming style  accessor methods that get and set values with each other as well as with methods that create and so initialise a new record.

### Example answer

Accessor methods that get and set values along with create methods are all just normal methods but used to access the fields of a record. They are used to provide a layer of abstraction from the actual implementation of the underlying record track, allowing that implementation (ie the underlying record) to be changed and then only the relevant methods need to also be changed, rather than lots of places throughout the program (wherever the changed fields are accessed). This is important to make programs easy to maintain ie change when bugs are found or when the program needs to be modified to do new things.

So for example if you have create a new Car type the get, set and create methods become the only ways that record values of that type are accessed or changed. You only ever create a car record using the create method, you only ever get the field values with the get method and only change the fields with the set methods...never directly.

The difference is then that get methods are used to access the values of the fields (or some value derived from them) whereas set methods are used to set the values of the fields. This means that set methods need to be provided with extra arguments - the new values for the fields, whereas get methods do not. They just need to be given as argument the record value to be accessed. A create method does not need to be given a record (eg Car) value. It's job is to return a new one (like a set method, though that updates a given one rarther than create one from scratch.)

Set methods have as aim to change the record value and so after they are called the record is changed (either by the new value being returned or otherwise). Get methods on the other hand are not intended to change anything: they access the value, leaving the record alone. The value returned to them is a field value.

Both set and create methods, put values in the fields of a record.

For example the following would be get, set and create methods for the car records defined above

```java
public static String getCarMake(Car c)
{
   return c.carmake;
}
```

This extracts and returns the carmake field of the given record.

```java
public static Car setCarMake(Car c, String newmake)
{
   c.carmake = newmake;

   return c;
}
```
This is given a new value to store in the carmake field and returns the updated car.

```java
public static Car createCar(String make, String reg, double prem)
{
   Car c = new Car();

   c.carmake = make;
   c.carregistration = reg;
   c.premium = prem;

   return c;
}
```

Notice how all return values but the get methods return a field value, whereas the others return awhole car. The get and set methods both need to be given the Car in question whereas the create method doesn't have a car as argument as its job is to create one. Once written all can be used to provide a layer of abstraction. With them, no where else in the program needs to use the dot notation or actual field names. Any time a car value is to be used one of the get, set or create methods can be called to do the job.

````java
/* ***************************************************
AUTHOR STUDENT ID 123456789
This program gives the deadly score of given animals.
Note later in the module we will see better ways of doing this!
*/
import java.util.*;

class DeadlyAnimals
{
    // Ask for an animal and print its deadly details.
    //
    public static void main(String [] p)
    {
        String animal = ""; // the chosen animal
        int score = 0;  // the chosen animals score
        boolean known = false; // Is the named animal known or not
        String kind; // the chosen animal
        
        Animal tiger = new Animal();
        Animal dog = new Animal();
        Animal kitten = new Animal();
        
        // Set the details of the known animals
        //
        tiger = setAnimalKind(tiger, "Tiger");
        tiger = setDeadlyScore(tiger, 9);
        
        kitten = setAnimalKind(kitten, "Kitten");
        kitten = setDeadlyScore(kitten, 0);
        
        dog = setAnimalKind(dog, "Dog");
        dog = setDeadlyScore(dog, 5);
        
        kind = inputString("Give me a kind of animal and I will give back its deadly score");
        
        // Extract information for chosen animal
        //
        if (kind.equals("Tiger") || kind.equals("tiger") || kind.equals("TIGER"))
        {
           animal = getAnimalKind(tiger);
           score = getDeadlyScore(tiger);
           known = true;
        }
        else if (kind.equals("Dog")|| kind.equals("dog") || kind.equals("DOG"))
        {
           animal = getAnimalKind(dog);
           score = getDeadlyScore(dog);
           known = true;
        }
        else if (kind.equals("Kitten") || kind.equals("kitten") || kind.equals("KITTEN"))
        {
           animal = getAnimalKind(kitten);
           score = getDeadlyScore(kitten);
           known = true;
        } 
        else
        {
           System.out.println("I don't know that animal");
        }
        
        // Print appropriate message given choice of animal
        //
        if (known)
        {
          printAnimalMessage(animal, score);
        }
        
        return;
    } // END main

    // Print an appropriate message about the animal
    //
    public static void printAnimalMessage(String animal, int score)
    {
        System.out.println("The " + animal + " has deadly score " + score);
        return; 
    } // END printAnimalMessage

    // Getter methods for Animal record type
    // Return the kind of animal from the record
    //
    public static String getAnimalKind(Animal a)
    {
       return a.animalkind;
    } // END getAnimalKind
    
    // Return the deadly score /10 from an animal record
    //
    public static int getDeadlyScore(Animal a)
    {
        return a.deadlyscore;
    } // END getDeadlyScore
    
    // Setter methods for Animal record type
    // Set the id of a student returning the updated record
    //
    public static Animal setAnimalKind(Animal a, String newkind)
    {
        a.animalkind = newkind;
        return a; 
    } // END setAnimalKind
    
    // Set the deadly score of an animal returning the updated record
    //
    public static Animal setDeadlyScore(Animal a, int newanimalscore)
    {
        a.deadlyscore = newanimalscore;
        return a; 
    } // END setDeadlyScore
    
    // A method to input Strings
    //
    public static String inputString(String message)
    {
        Scanner scanner = new Scanner(System.in);
        System.out.println(message);
        return scanner.nextLine();
  } // END inputString
    
} //END class DeadlyAnimals
````
