## Question 1 \[10 marks\]

**State** how many times the letter p is printed in each of the following examples. 
**Justify** your answer by completing the dry run tables.

i) \[2 marks\]  
```
for(int i = 1; i < 4; i++)  \\ L1
{  
    System.out.println(”p”); \\ L2
}
```

ii) \[2 marks\]
```
for(int i = 1; i < 1; i++)   \\ L1
{  
    System.out.println("p"); \\ L2
}
```

iii) \[3 marks\]  
```
for(int i = 1; i <= 2; i++)        \\ L1
    for(int j = 1; j <= 2; j++)    \\ L2  
    {  
        System.out.println("p");   \\ L3 
    }  
}  
```

iv) \[3 marks\]  
```
for(int i = 1; i <= 4; i++)        \\ L1
    for(int j = 1; j <= i; j++)    \\ L2  
    {  
        System.out.println("p");   \\ L3  
    }  
}
```

### Answers

i) Answer 3.

Justification:

The counter counts ```1```, ```2```, ```3```. When it gets to ```4``` ```i``` is no longer less than ```4``` so it stops.

ii) Answer 0.  

Justification:


```i``` starts at ```1``` but this is already not less than ```1``` so it never goes into the loop body.

iii) Answer 4. 

Justification:

Consider the inner loop alone. It executes 4 times as it counts ```1```, ```2```. Only when it gets to ```3``` does the test become ```false``` and stop. Now consider the inner loop as a single thing. The outer loop says to do it 2 times (again counting ```1```, ```2```). Therefore the inner loop prints 2 times and this is done 2 times over, giving a total of 4.

iv) Answer 10.  

Justification:

The outer loop executes 4 times as for the reasoning above, counting ```i``` as ```1```, ```2```, ```3```, ```4```. Now what the inner loop does depends on ```i``` as it counts up to ```i```. When ```i``` is ```1```, it prints just once as ```j``` is then more than ```i```. When ```i``` is ```2```, ```j``` counts ```1```, ```2```. When ```i``` is ```3```, ```j``` counts ```1```, ```2```, ```3``` and when ```i``` is ```4```, ```j``` counts ```1```, ```2```, ```3```, ```4```. Each count results in the character being printed. Therefore the first loop results in 1 print, the second 2 prints, the third 3 prints and the fourth 4 prints. The total is 1+2+3+4=10

## Question 2 \[10 marks \]

The following code fragments contain errors. For each **state** what the error is and whether the compiler would detect it **justifying** your answer.  

i) \[3 marks\]  
```java
// write Hello 10 times
//
for(int i=0; i < 10; i++);
{  
    System.out.println("Hello");  
}
```

ii) \[3 marks\]  
```java
// write numbers from 1 to 10 
//
for(int i=1; i < 10; i++)  
{  
    System.out.println(i);
}

```

iii) \[2 marks\]  
```java
// write numbers from 1 to 4 
//
for(int i=1; i <= 4; j++)  
{  
    System.out.println(i);
}
```


iv) \[2 marks\]  
```java
// write numbers from 1 to 7 
//
for(i=1; i <= 7; i++)  
{  
    System.out.println(i);
}
```


## Feedback on the kind of answer expected

*Justification is again asked for. Answers would get a maximum of one mark if only the error was pointed out.*


## Example good answers

i)

There should not be a semicolon at the end of the for line. This would not be detected by the compiler. Instead it would treat the program as though it was a loop with no body followed by a single line that has nothing to do with the loop. That is the code is logically equivalent to the following (though would be very slightly slower to execute unless the compiler ignores the loop that does nothing.
```java
// write Hello once
//
System.out.println("Hello");
```

ii)  

The problem is that the loop test is wrong thing. 

This would not be detected by the compiler. It will both compile and run, but do the . 

However it will write numbers from 1 to 9 not 1 to 10 as the test says stop doing the loop once ```i``` is 10. Thus it goes into the loop when ```i``` is ```9```, printing 9, increments ```i``` to ```10``` but then the test becomes ```false``` so it does not execute the body again. If the aim is to print the numbers from 1 to 10 then the following code is needed instead with a different loop condition.

```java
// write numbers from 1 to 10
//
for(int i=1; i <= 10; i++)
{
    System.out.println(i); 
}
```

iii)

The problem here is that the wrong variable is incremented. It should be ```i++```, incrementing ```i```, but instead increments ```j```. 

This would be detected by the compiler as long as the variable ```j``` has not been previously declared within the method. In that case the compiler will complain that variable ```j``` is not declared and so no runnable program will result. If it has been declared and initialised  then the code would become a non-terminating loop as i never changes so the condition is always false.

iv)  

The problem here is that the loop counter variable ```i``` is not declared. The initialisation  should include a declaration so be ```int i = 0;```

This would  be detected by the compiler as long as the variable ```i``` has not been previously declared within the method. In that case the compiler will complain that variable ```i``` is not declared and so no runnable program will result. If it has been declared then the code will run correctly. However, it is very bad style to declare a loop counter outside the loop> It is important because it makes the loop understandable just looking at the loop. It avoids the prospects of mistakes where the loop counter is mistakenly used for something else outside the loop.

## Question 3 [10 marks]


**Write** a Java program that uses a for loop to print the n times table, where n is a number supplied
by the user. An example run might be:

```Which times table should I print?``` **```2```**  
```1x2=2 2x2=4 3x2=6 4x2=8 5x2=10```




### Feedback on the kind of answer expected

*The following table gives an indicative idea of what marks an answer deserves:*

**A: 7-10 marks:**

- Logically correct program. It does the right thing.
- Greater marks for doing it elegantly...comments,
- efficient algorithm, methods etc.

**B: 6 marks:**

- Does something, but not quite the right thing.
- Contains at least one logical error or multiple syntax errors, but basically shown can solve problem.

**C: 5 marks:**

- Does something, but not quite the right thing.
- Contains more logical errors, but basically shown can solve problem.


**D: 4 marks:**

- More major errors but evidence that can write a program using appropriate
- constructs if put in front of a compiler - though it may take time.

**F/G: 0-3 marks:**

- Fragmentary code answer - eg rote-learnt bits of code.
- No clear evidence can write a program to solve a problem.

### Example answer

````java
/* 
   AUTHOR STUDENT ID 123456789

   Prints a requested multiplication table 
*/
import java.util.*;

class TimesTable
{
    public static void main(String [] param)
    {
        int table = whichTable();
        String outputText = generateTimeTable(table);
        printTable(outputText);
        return;
    } // END main
    
    // Determine the table to print
    //
    public static int whichTable()
    {
        int table = inputInt("Which table should I print?");
        
        return table;
    } // END whichtable

    // Generate a string holding the result that needs printing
    //
    public static String generateTimeTable(int table)
    {
        String outputText = "";
        
        for(int i = 1; i<=5; i++)
        {
            outputText = outputText + i + "x" + table + "=" + (i*table) + "\t";
        }
        
        return outputText;
    } // END generateTimeTable
    
    // Print the final result
    //
    public static void printTable(String message)
    {
        System.out.println(message);
        return; 
    } // END printTable
    
    // Input an integer
    //
    public static int inputInt(String message)
    {
        return Integer.parseInt(input(message));
    } // END inputInt
    
    // Input strings
    //
    public static String input(String message)
    {
        Scanner scanner = new Scanner(System.in);
        String ans;
        
        System.out.println(message);
        ans = scanner.nextLine();
        
        return ans;
  } // END input
    
} // END class TimesTable
````
