## Question 1 \[10 marks\]

*State* whether each of the following are good or bad choices for variable names, *justifying* your answers.


i) ```t```  

ii) ```class```  

iii) ```hair_colour```  

iv) ```hairColour1```

v) ```4cast```  

vi) ```timeOfDay?```

vii) ```timeofday```

viii) ```one```

ix) ```123```

x) ```time of day```

### Feedback on the kind of answer expected

*Note that this question asks for justifications. To get the marks you need to say more than just "good" or "bad" but explain why they are good or bad too. There are no marks for making 50-50 guesses - a justification shows you understand why. As it is only 1 mark each though the justifications will be brief.*



## Example answer

i) ```t```   
BAD. Single letter identifiers give little information about their use.

ii) ```class```  
BAD. class is a reserved word (a keyword) - it already means something in Java so cannot be used as a variable as if it were allowed it would not be clear whether it referred to the variable or to the keyword.

iii) ```hair_colour```  
GOOD. Underscores are allowed and the name used is fairly descriptive so making the program more readable.

iv) ```hairColour1```  
GOOD. As above. Digits are allowed at the end of identifiers - just not at the start. Camel case is a good way to split words.

v) ```4cast```  
BAD digits are not allowed to start an identifier and it is not very descriptive anyway.

vi) ```timeOfDay?```  
BAD. An identifier cannot contain symbol characters like a question mark.

ii) ```timeofday```
BAD. (but not too bad) Rather than run words together it is better to have some visual way to separate them.

viii) ```one```
BAD. If used to store the number 1 then some name that shows what the 1 means, like ```STARTING_POINT``` would be better.

ix) ```123```
BAD. Numbers cannot be used as variable names.

x) ```time of day```
BAD. Spaces cannot be used in variable names as they are used as white space to distinguish different syntactic units. They therefore can't be used as a separator in a variable name.

## Question 2 \[10 marks\]

**Compare and Contrast** the following fragments of code.

i)  
```
t = a;  
a = b;  
b = t;
```
ii)
```
c = d;  
d = c;
```

### Feedback on the kind of answer expected

*This question does not ask just for an explanation but for you to explain the similarities and differences of the two things. Roughly half the marks are for comparison points, half for contrasts. Maximum 4 marks (a bare pass) if it just explains code without an explicit attempt to give similarities and differences. For high marks you should show understanding of overall effects rather than just a line by line comparison.*

*For a 10 mark question expect to be giving 2-3 comparison and 2-3 contrast points well explained using examples for high marks*

Note how the answer below compares point by point and then contrasted point by point in an integrated way. It does not just give a list of points about the first code and then a separate list of points about the second code fragment. We have also bolded words like 'Both` that are indicators of a comparison and **on the other hand** indicators of a contrast point being made.

Examples are used to illustrate comparison and contrast points, not just individual points about one or the other thing being compared / contrasted.


### Example answer

**Both** these pieces of code consist of a series of assignments to variables (of the form ```x = y``` for some x and y, ie that store a value in a variable for later use. Here the values copied themselves come from variables **in both cases**. They **both** thus copy data from one place to another. The new values of variables **in both cases** are just copies of values stored in existing variables (the right-hand sides of the assignments are expressions consisting soley of variables). So both ```java t = a``` and ```java c = d``` are for example taking a copy of a value from a variable (```a``` and ```d```) and putting it in another (```t``` and ```c```.

**Both** are probably attempts to write code to swap values between two variables. **In either case** the variables must all be the same type and this could be any basic type. 

In **both** cases the variable names are poorly chosen as they have single letter names like ```a```, and ```c```.

A trivial **difference** is that the variables concerned are named differently (```a```, ```b``` and ```t``` versus ```c``` and ```d```).

The first is correct code to do a swap of values between two variables, **whereas** the second does not work. The first swaps the value from ```a``` into ```b```, and the value of ```b``` into ```a```, leaving ```t``` which was acting as temporary storage space, with the original value of ```a```. The second fragment **on the other hand** tries to do the swap without an extra variable as a temporary storage space. The result is that unlike the first example both variables end up with the same original value of ```d``` as the value originally in ```c``` is destroyed (and lost for good) by the first assignment.

## Question 3 \[10 marks\]


**Write** a Java program that converts a length of time given in days to a length of time in minutes. For example 1 day would be converted to 1440 minutes (as 1x60x24=1440).


### Outline marking scheme

*The following table gives an indicative idea of what marks an answer deserves:*

**A: 7-10 marks:**

- Logically correct program. It does the right thing.
- Greater marks for doing it elegantly with good style...comments, etc.
- efficient algorithm, good method decomposition, etc.

**B: 6 marks:**

- Does something, but not quite the right thing.
- Contains at least one logical error or multiple syntax errors, but basically shown can solve problem.
- Some style problems

**C: 5 marks:**

- Does something, but not quite the right thing.
- Contains more logical errors, but basically shown can solve problem.
- very poor style

**D: 4 marks:**

- Has shown can use key programming constructs needed for the question
- More major errors but evidence that can write a program using appropriate constructs if put in front of a compiler, though it may take time.

**F/G: 0-3 marks:**

- Fragmentary code answer - eg rote-learnt bits of code.
- No clear evidence can write a program to solve a problem.

### Feedback on the kind of answer expected

The program below would probably get a borderline A in that it is correct and uses methods. However the decomposition is poor with only one method split off. There are clear sub-tasks of calculating the minutes (or even daysToHours and HoursToMinutes and printing the answer. These could have been made methods. Also a more general input method could have been used. We will cover more on how to do this well  in later units, however.

Style is fairly good in use of comments, indentation, blank lines, etc. It is over commented and the variable names could have eg underscores separating the words. An answer that is hard for an expert (the marker) to uickly read and understand is unlikely to  get high marks. Good style contributes to this.

Note also that by the mid term test you will be expected to be using if statements and for loops (to come) too.

### Example answer

````java
/* *****************************************
AUTHOR Paul Curzon
STUDENT ID: 123456789

Converts a length of time given
in days to a length of time in minutes.
For example 1 day would be
converted to 1440 minutes (as 1x60x24=1440).
******************************************** */

import java.util.Scanner; // Needed to make Scanner available

class DaysMinutes
{
    public static void main(String[] param)
    {
        convertDaysToMinutes();
        return;
    } // END main
    
    /* *************************************************** */
    // Get a number of days from the user and print
    // the corresponding number of minutes.
    // 
    public static void convertDaysToMinutes()
    {
        // Declare variables
        //
        int days;     // a length of time in days
        int minutes;  // the same length of time in minutes
        
        // constant multipliers for hours in a day and minutes in an hour
        final int HOURS_IN_DAY = 24;
        final int MINS_IN_HOUR = 60;
        String answertext;  // construct the text to give the answer back
        
        // Get a number of days from the user
        //
        days = inputTimeInterval();
        
        // Now do the calculation
        // multiply minutes in an hour by hours in a day
        //
        minutes = days *  MINS_IN_HOUR * HOURS_IN_DAY;
        answertext = days + " days is "+ minutes + " minutes";
        
        // Finally give the user the answer
        //
        System.out.println(answertext);
        return;
    } // END convertDaysToMinutes
    
    /* *************************************************** */
    // This method gets a time interval in days from the user 
    // and returns it
    //
    public static int inputTimeInterval()
    {
       Scanner scanner = new Scanner(System.in);
       int t;
       System.out.println("What is the time interval in days?");
       t = Integer.parseInt(scanner.nextLine());
       return t;
    } // END inputTimeInterval
    
} // END class DaysMinutes
````
