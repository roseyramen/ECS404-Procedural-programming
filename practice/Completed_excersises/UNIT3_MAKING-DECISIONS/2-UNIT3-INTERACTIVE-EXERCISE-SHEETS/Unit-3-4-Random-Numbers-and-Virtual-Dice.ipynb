## Introduction
Many programs need randomness. Games are the obvious example. If you played a chess program and it always made the same moves so was totally predictable it would ultimately be easy to win.

However, programs do exactly what their instructions tell them to, so this implies that with the same inputs they will do exactly the same thing.

This apparent paradox is broken using random number generators. Essentially programming languages provide a method that gives back different numbers each time it is called. This could for example, be based on how many nanoseconds the machine has been running.

### Solution to Exercise 1<a id="Solution1"></a>

The method as a whole prints a single random number between 0 and 2 inclusive. Each time the method is called it gives a new random number in that range.

The line:
```
Random stream_of_random_numbers = new Random();
```    
creates a new stream of random numbers stored in a variable called ```stream_of_random_numbers```
It is just a variable so you could have called it anything.

You can imagine it as an infinite list of numbers. Those numbers can be 'plucked' off the list one at a time when needed. 

The method nextInt plucks the next number off the list so 
```
stream_of_random_numbers.nextInt(3);
```
returns a single random number (removing it from the front of the list so that next time it is called another random number is released).

Why is the argument 3 needed? The numbers in the list are actually real numbers between 0 and 2. This means they have to be converted to the range needed. If you want a normal 6-sided dice you need numbers between 1 and 6 for example. The number 3 gives the range. It means the numbers returned are integers between 0 and 3. It is not as you might have thought giving the upper number but the number of possible different numbers that will be generated. By putting 3 there you are saying you want 1 of 3 different numbers to be generated starting from 0 (so from {0,1,2})

You may have incorrectly guessed that it would return numbers 1 or 2. In fact it always produces a number from 0 up to one less than the number given.

In the method, the number returned is stored in the integer variable we called ```random_number```. Again it is just a variable and we could have called it anything. Always call the variable something descriptive.

```
int random_number = stream_of_random_numbers.nextInt(3);
```

### Solution to Exercise 3<a id="Solution3"></a>

You just need to add 1 to the random number returned. That turns any random number 0 into 1, 1 into 2 and so on up to 5 into 6, giving the right range of numbers.

````java
// This prints a random number between 1 and 6 inclusive

public static void generateRandomNumber1to6()
{
    Random stream_of_random_numbers = new Random();
    
    int random_number0to5 = stream_of_random_numbers.nextInt(6);
    int random_number1to6 = random_number0to5 + 1;
         
    System.out.println(random_number1to6);
    
    return;
} // END generateRandomNumber1to6
````

### Solution to Exercise 6<a id="Solution6"></a>

The code does the equivalent of a coin toss returning one of two possible strings at random - either Heads or TAILS.

````java
// Toss a coin virtually retirning either the String HEADS or TAILS
//
public static String coinToss()
{
    String toss;
    Random stream_of_random_numbers = new Random();
    
    int random0to1 = stream_of_random_numbers.nextInt(2);
    
    if (random0to1 == 0)
        toss = "HEADS";
    else
        toss = "TAILS"; 
        
    return toss;
}
````

````java
// Decide at random what a batsman hits in a cricket game - FOUR, SIX or OUT
//
public static String cricketHit()
{
    String hit;
    Random stream_of_random_numbers = new Random();
    
    int random0to2 = stream_of_random_numbers.nextInt(3);
    
    if (random0to2 == 0)
        hit = "FOUR";
    else if (random0to2 == 1)
        hit = "SIX"; 
    else
        hit = "OUT"; 
        
    return hit;
}
````

### Solution to Exercise 8<a id="Solution8"></a>

- Random number generators can be made available to a program by loading in the library java.util.Random.
- First you must create a stream (essentially an infinite list) of random numbers 
    - they actually are real numbers between 0 and 1
    - this is done using eg
    - ```Random r = new Random();```
    - This creates a new variable called r that holds a stream of random numbers
    - It is very much like creating an input stream for doing keyboard input
- Once you have a stream you can pull numbers off it one at a time using the method nextInt(...)
    - It takes an integer argument which gives the number of possibilities for the random number returned.
    - So if you give it the argument 4 as in nextInt(4) then it will return 1 of 4 values from 0, 1, 2, 3, 4
    - to call it you need to link it to the particular stream of random numbers by putting the stream variable name and a . before it
    - eg ```r.nextInt(4)```
    - This returns an integer so it can be used or stored as any other integer. Most likely it it stored in a variable for later use.
    - The integers returned can be converted into other values eg String text just using a conversion method with a big if statement.
