## Introduction: What is sorting?
How do you put things in to order? You need a sort algorithm. It is an organised way that guarantees to put data into order. Bubble sort is one very simple way of sorting data. Here we are going to focus on something more fundamental - what do we mean by the **order** we are sorting things in to as well as what it actually means to sort different types of value into order.

There are many meanings of the verb ‘to sort’. It could mean to fix things, it could mean to put things in to groups. To a computer scientist ‘sorting’ means something very specific. It means taking a series of pieces of data and putting them into a set order, one after the other, either from smallest to largest or from largest to smallest. 

For example, sorting numbers means putting them in to numerical order. If we are given the numbers:
5, 7, 2, 99, 4
then after sorting them in to ascending order we have the list:
2, 4, 5, 7, 99
If we were to sort them into descending order then, instead, we would get the list:
99, 7, 5, 4, 2

````java
public boolean lessThanMedalCountCountry(MedalCount mc1, MedalCount mc2)
{
    if ( lessThanString2(mc1.country, mc2.country) )
        return true;
    else
        return false;
} // END lessThanMedalCountCountry
````

### Solution to Exercise 6<a id="Solution6"></a>

````java
public boolean lessThanString(String s1, String s2)
{
    int shortestLength;
    int length1 = s1.length();
    int length2 = s2.length();
    
    // First find the length of the shortest of the two strings
    if (length1 < length2)
        
        shortestLength = length1;
    else
        shortestLength = length2;
    
    // Then check them character by character up to that point
    for (int i = 0; i< shortestLength; i++)
    {
        if (s1.charAt(i) < s2.charAt(i))
            return true;
        else if (s1.charAt(i) > s2.charAt(i))
            return false;
    }
    
    // if we get to this point they or the same or one is shorter than the other
    if (length1 == length2)
        return false; // They are the same so not less
    else if (length1 < length2)
        return true; // s1 is shortest so comes first
    else
        return false; // s2 is shortest so comes first

} // END lessThanString
````

### Solution to Exercise 12<a id="Solution12"></a>


````java
// Check if the first country is higher in the medal table than the second
// Based on who has most golds, then most silvers, then most bronzes.
//
public boolean greaterThanMedals(Medals country1, Medals country2)
{
    if (country1.gold > country2.gold)
        return true;
    else if (country1.gold < country2.gold)
        return false;
    else if (country1.silver > country2.silver) // golds must be equal by this point
        return true;
    else if (country1.silver < country2.silver)
        return false;
    else if (country1.bronze > country2.bronze) // golds and silvermust be equal by this point
        return true;
    else
        return false;
    
} // END greaterThanMedals
````

````java
Medals country1 = new Medals();
country1.country = "China";
country1.gold = 38;
country1.silver = 32;
country1.bronze = 17;

Medals country2 = new Medals();
country2.country = "USA";
country2.gold = 39;
country2.silver = 41;
country2.bronze = 33;

Medals country3 = new Medals();
country3.country = "Mars";
country3.gold = 2;
country3.silver = 60;
country3.bronze = 12;

Medals country4 = new Medals();
country4.country = "Venus";
country4.gold = 2;
country4.silver = 15;
country4.bronze = 277;

System.out.println(greaterThanMedals(country1, country2));
System.out.println(greaterThanMedals(country1, country3));
System.out.println(greaterThanMedals(country1, country4));
System.out.println(greaterThanMedals(country1, country1));
````

### Solution to Exercise 13: Summary

Sorting is to put things in to order and to do that you need an operator that for the values you want to sort tells you which comes before which. It should be able to do this for any pair. Given two values it returns true if they are in the right order and false if they are in the wrong order. < and > (for integers and characters) or a method based on compareTo (for strings) can be used to do this. When we define our own types (eg records) we must defined the order of the records our self by defining a method that does this ordering. 

Once we have a method for ordering values of a given type we can then use it in a method that implements a sort algorithm to sort arrays of such values (eg sort arrays of Strings or arrays of medal counts i.e. medal tables). Essentially the same method can then be used to sort anything, just by swapping in an appropriate comparison method.
