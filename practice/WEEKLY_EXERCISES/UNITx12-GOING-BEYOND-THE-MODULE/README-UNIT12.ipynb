### Challenge 3: Functional programming

This builds on recursion. Redo your mini project as a pure functional program.

Learn about functional programming and improve your understanding of recursion: re-implement all the exercises in Java but in the style of a functional program.

The following are the main differences between procedural programming and functional programming.

- **Functional programs do not use assignment.** Variables can be given initial values but then those values are never changed. In effect all variables are declared as constants (ie final). This is called variable binding in functional programs.
  
- **Functional programs do not have sequencing.** They use function call, passing the values returned from a function on to the next function call as arguments to do one thing after another and to bind values to function arguments.

- **Functional programs also do not use loops.** Instead of constructs like for and while they use function call and specifically recursion to do the same thing over and over again.

- **Functional programs do not have methods that have side effects.** Values that are the result of a method are returned. There is no use of call by reference style side effects.
1) Try doing the exercises and ultimately the mini-project again but following these restrictions.

You are programming in a form closer to pure logic, which it turns out is the basis of the way specifications for programs are written and opens up possibilities of proving mathematically that programs work. The restrictions mean you can repeatedly substitute arguments in for the names of formal parameters in definitions as a way of dry running a functional program.

2) Languages that enforce restrictions like this include Haskell, Standard ML and Pure Lisp. They are called functional programming languages. Once you have done it in Java find out more about these other languages and see if you can convert your program into their syntax and get it to work.

3) Write a program to evaluate simple integer based functional programs by substitution as described above. Represent the bodies of method definitions as strings and do repeated string substitution on a string representing a call. When you get to a simple expression like 1+2 with only arithmetic operators, evaluate it directly. So for example starting with a call "wombat (5)" and definition "wombat (n) = n+1" You would replace "wombat(5)" by definition body with 5 substituted everywhere in it for n i.e.  "5+1" . That would then evaluate to 6. NB See the expression parser from the recursion unit as a way of parsing and then evaluating simple arithmetic expressions.

### Challenge 9: Naked Mole-rats (Hard)

Only the queen of a naked mole-rat colony has pups. They live amazingly long lives and are highly resistant to cancer! Here is a programming puzzle about a fictional colony...

A one-year old queen forms a new colony with a single male. Each year they have a new litter of pups. The number of pups for this queen just so happen to follow a special pattern: 2 pups in the first year, then 3 the next, 5 the next, 8 the next, and so on. This queen dies before any of her pups or the original male. 

Write an efficient recursive program to work out the size of her colony after each year if the pattern is followed until she dies (exhausted).

Note the emphasis here is on it being efficient...which ultimately means I want  a divide and conquer algorithm, but first get a simple recursive algorithm to work and use it as a specification to test more efficient answers against automatically.
