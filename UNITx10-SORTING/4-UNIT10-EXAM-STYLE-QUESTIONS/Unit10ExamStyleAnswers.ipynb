## Question 1 \[10 marks\]

i) 

**Explain** how with a simple change, just to the inner loop, the following algorithm to sort data can be improved, **justifying** your answer.
```java
for(int pass = 0; pass <= array.length-2; pass++)
   for(int i = 0; i <= array.length-2; i++)
   {
     if (a[i] < a[i+1]))
     {
           swap(a, i, i+1)
     }
   }
```
You may assume that swaps the values in an array at the given positions.

ii)

**Illustrate** your answer by showing what happens when each algorithm is applied to an array {4,3,2,1} by showing how the order of the array changes after each iteration.

### Example answer

i)

The algorithm as  given runs to the end of the array on every pass. However, on each pass we compare pairs of values swapping them if out of order and so moving the largest towards the end. In this way we take the largest found so far with us and that means at the end of the pass, the largest value is moved to its final position at the end. That means it no longer makes sense to compare that element on later passes as it is in the right place and is not going to move. We can stop the next pass a place earlier. The same reasoning applies to each pass. The second pass will put the second largest value in the right place second from the end, and so on. So on each pass we can reduce how far we go by one position.

This can be done just by changing the condition on the inner loop. The outer loop counts the pass from 0. On pass 0 it is fine, but on pass 1 we can subtract 1 from the value, on pass 2, subtract 2 and so on. This means in fact we just need to subtract pass itself giving a test  
`i <= array.length - 2 - pass`

## Question 2 \[10 marks\]

**Compare and contrast**:  

i) bubble sort, and 

ii) merge sort

### Example answer


Both bubble sort and merge sort are sort algorithms - they both aim to put data in order. In both cases, this could be numerical, alphabetical, or in fact any order that is defied by a less than relation for both algorithms. It could also be ascending or descending order.

However, the algorithms take completely different approaches one slow the other fast. Bubble sort is a relatively slow algorithm whereas merge sort is one of the fastest.

Bubble sort works by a series of passes of the data comparing adjacent data as it runs from one end of the array to the other, whereas merge sort does not necessarily compare adjacent data in the array. 

Each pass in Bubble sort moves an element a little way towards its final destination. For example when sorting 4 elements, initially in the order 9, 5, 7, 2. It would compare 9 and 5 and swap them, then the 9 and 7 and swap them and then the 9 and 2 and then swap them.

Each pass only guarantees one piece of data in the right place (in the above example the first pass guaranteees the 9 has moved to the right place). The algorithm is relatively slow and inefficient as a result. Given n pieces of data to sort it needs n-1 passes of the data each doing in the order of n-1 comparisons or around n/2 if more efficient versions are used. Sorting 8 pieces of data takes 7 passes each doing at most 7 comparisons. An efficient version does 7 comparisons, then 6 comparisons, ... then 2 comparisons, then 1 comparison, so in total 7+6+ ...+2+1.

By contrast merge sort works by divide and conquer and so is a very fast algorithm. The problem is broken into similar but smaller problems of half the size. In particular,  it splits the data in half. Each is sorted in the same way using merge sort also splitting. The halves come back sorted and are then merged using a merge algorithm in to the right order. It thus does not work based on swaps of pairs in the way bubble sort does though still on comparing pairs and taking the smallest each time as part of the merge algorithm. Comparisons are done when merging the halves back together. With 8 pieces of data there are 4 pairs so 4 comparisons done to put the pairs in order, those 4 pairs become 2 groups of 4 needing 6 comparisons, Those 2 groups of 4 are put into order with 7 comparisons.  This is 7+6+4 comparisons. The more data the bigger the speed difference.

Bubble sort is most naturally described iteratively, whereas merges sort, as it is a divide and conquer algorithm, is naturally described recursively, though implementations may be iterative for increased efficiency.

As a result of the speed difference, bubble sort would only be used to sort small arrays as it would not be able to do it in a reasonable time, whereas merge sort could be used to sort very large amounts of data.

## Question 3 [10 marks]

**Discuss** the truth or otherwise of the statement "Searching is faster if data is sorted".

### Feedback on the kind of answer expected

*A question like this asking for a discussion of the truth or otherwise is asking for an essay-style answer that has a clear introduction, body and conclusion and that looks at arguments for both sides before drawing a conclusion.*

### Example answer

In general if data is sorted in to order then more efficient algorithms such as binary search can be used to search for elements. If data is not sorted then either you must sort first or use an inefficient algorithm (most obviously linear search) checking every element in turn. This is significantly slower. For example with a million pieces of data you might take at worst 20 comparisons to find the thing you are looking for. With linear search on average you would take half a million comparisons to do the same job and a million in the worst case. It is thus massively slower.

However, if the data was by chance near the start you could still find it much quicker. Linear search finds elements that happen to be in the first position in a single comparison. If you had data where it was known that the things searched for were likely to be near the start then linear search might in practice be the best option, though this seems likely to be a fairly rare situation.

Also there is a trade off. If you have to sort first then that is much slower than searching, so sorting data to search it once would end up slower overall than just using linear search straight away. It is only worth sorting data if you are going to search it repeatedly.

Finally, there are other ways to organise data, in addition to sorting it into a linear order, that allow for fast search algorithms. For example, approaches such as table lookup where data is dropped in to buckets one for each possible value, and you go straight to the bucket to find it, can be even faster. There are trade-offs though. This is only practical for small data ranges. It also potentially uses a lot of storage space to store the lookup table (though there are ways round that).

In conclusion, in general it is true that faster search algorithms become available if data is sorted, but in some circumstances searching my still be faster even if it is not sorted. There are also other ways than sorting to give opportunities for fast searching.
