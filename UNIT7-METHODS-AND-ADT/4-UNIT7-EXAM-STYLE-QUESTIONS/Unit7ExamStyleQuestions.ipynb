**Exam style questions are questions that are in a similar style to questions that you will have to answer in tests and the exam.** The idea is to give you a chance to practice early, as you do units (revision should NOT be last minute but something you do week by week).  

Questions linked to each unit include questions focused on the material from that unit, though on the exam or a test a question could cover material from multiple units. Programming questions in particular are likely to involve a lot more for the marks than in the early unit questions.

As they are for practice, we have put them here in JHUB so you can easily check things after answering or that you are unsure of by running programs and fragments. However, your aim is to be able to answer immediately and correctly from your own knowledge, understanding and skill. You ultimately should not need to get such help from tools like JHUB, so do not use it as a crutch, use it to learn. It is far quicker to answer questions without such help and in tests you will  likely run out of time if you can't.

You could have to do similar questions in the exam or tests **in invigilated conditions on paper**, so practice doing that, especially writing programs. (And in job interviews it is common to ask people to write programs on the spot without any tool help). You will become a far better programmer generally if you practice doing this too.

**All answers in assessed tests and exams MUST BE WRITTEN IN YOUR OWN WORDS.** Practice doing that.

### Doing it again

Once you have answered the questions, read the answers carefully (link at end) and compare them with what you wrote - they are there for feedback, but feedback only helps if you use it actively. Then try and answer the questions again improving your answer. Always write in your own words though. There are no marks for just memorising my words. You have to show you can do it.

You can add extra boxes if you wish with notes about things you have learnt, issues that came up you want to remember, etc - eg to avoid making similar mistakes again in future.
