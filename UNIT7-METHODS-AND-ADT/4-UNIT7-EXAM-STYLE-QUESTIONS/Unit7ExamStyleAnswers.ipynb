## Question 1 \[10 marks\]

**Compare and contrast** the following, **illustrating** your answer with examples:  

i) method call  

ii) function call 

iii) procedure call

### Example answer

A method is the general term used to mean both functions and procedures. A method call is either a function call or a procedure call. A method call (so both a procedure call and a function call) is when a method definition is executed as part of a program running. It is a way of indicating when named, decomposed parts of a program should be executed. This is done by giving the name of the method followed by parentheses (for both procedure or function call). For example,
```java
method1(); 
```
is a call to a method called method1.

All methods can be defined to take arguments (whether a function or a procedure) - information to pass into the method to do its job. For example,
```java
method2("Paul", 100);
```
passes two arguments to method1, a String "Paul" and an integer 100. It will store these values locally and use them. 

The key difference is that a function call returns a value whereas a procedure call does not. A method in general could do either. It is whether a method call returns a value or not that determines whether the method *is* a procedure or a function. 

The point of a procedure is thus to do some task. Once finished the program moves on. It is a complete statement, so a whole command. 

For example, 
```java
System.out.println("Hello")
```
is a procedure call. Its purpose is to print a message as given in its argument. It is a whole statement and so appears in the same position in programs as other statements lime whole assignments.

The point of a function call on the other hand is to calculate a value, and once obtained further processing is then done with that value. A function call is an expression not a full statement like a procedure call. The value returned must therefore be saved somewhere (in a variable) or immediately passed on to some other operation. This is done using an assignment or embedding the call in a larger expression which itself is part of a larger statement.

So for example,
```java
ans = Integer.parseInt(anstext)
```
includes a function call to the function `Integer.parseInt`. The call returns a value computed from the argument passed and here this value is stored in variable ans for use later in the program. The function call is an expression and so appears in the same position as other expressions embedded inside statements: here on the right hand side of an assignment.

In summary, all are calls to methods and in all cases it is done in the same way by giving the name of the method followed by parentheses (possibly with arguments) as above. Method call is the general term, whereas the other two refer to the two kinds of method call based on whether a result is returned or not, and so whether the call is an expression or a statement.

## Question 3 \[10 marks\]

The Garage, Mile End Marvelous Motors sells used cars at discount prices. Given the full price of a car, the salespeople always knock off 10% plus give a further discount of the agreed price of any existing car that the customer traded-in. **Write** a Java method that calculates the value of a car given its full price and the trade-in price of the car being exchanged, both passed as arguments to the method. In addition write a main method that calls this method repeatedly with values input from the keyboard until the user indicates they wish to quit.


### Outline Marking Scheme

*The following table gives an indicative idea of what marks an answer deserves:*

**A: 7-10 marks:**

- Logically correct program. It does the right thing.
- Greater marks for doing it elegantly with good style...comments, etc.
- efficient algorithm, good method decomposition, etc.

**B: 6 marks:**

- Does something, but not quite the right thing.
- Contains at least one logical error or multiple syntax errors, but basically shown can solve problem.
- Some style problems

**C: 5 marks:**

- Does something, but not quite the right thing.
- Contains more logical errors, but basically shown can solve problem.
- very poor style

**D: 4 marks:**

- Has shown can use key programming constructs needed for the question
- More major errors but evidence that can write a program using appropriate constructs if put in front of a compiler, though it may take time.

**F/G: 0-3 marks:**

- Fragmentary code answer - eg rote-learnt bits of code.
- No clear evidence can write a program to solve a problem.

### Example answer

````java
/*
   AUTHOR: SID 123456789
   29 Oct 2023

   Calculate car prices based on a discount and trade-in price

*/
import java.util.*;

class CardDiscount
{
    public static void main(String [] param)
    {
        String cont = "y";
        
        while(isYes(cont))
        {
            int price, full_price, tradein_price;

            full_price = inputInt("What is the full price?");
            tradein_price = inputInt("What is the trade-in price?");
            
            price = calculatePrice(fullprice, tradeinprice);

            System.out.println("The discounted price is " + price + " UKP");

            cont = input("Continue (y/n)?");
        }
        
    return;
  } // END main
  
    // The price of a car is the full price less a 10% discount less the price 
    // agreed on any trade-in car
    //
    public static int calculatePrice (int full, int tradein)
    {
        int discount = full * 10 / 100;
        
        return (full - discount - tradein);   
    } // END calculatePrice

    public static boolean isYes(String answer)
    {
        boolean is_yes = answer.equals("y") || answer.equals("Y") ||
                        answer.equals("yes") || answer.equals("Yes") ||
                        answer.equals("YES");
        return is_yes;
    } // END isYes
 
    // Input Integer
    public static int inputInt(String message)
    {
        return Integer.parseInt(input(message));
    } // END inputInt

    // Input strings
    public static String input(String message)
    {
        Scanner scanner = new Scanner(System.in);
        String ans;

        System.out.println(message);
        ans = scanner.nextLine();
        return ans;
    } // END input

} // END class CardDiscount
````

## Question 4 \[10 marks\]

Using the example of a priority queue, a queue where entries leave in order of a priority given to them at the point when joined the queue, **explain** what is meant by an abstract data type and how methods can be used to create abstract data types. As part of your answer, **give** a specification of a priority queue as a series of method headers to **illustrate** your answer. You do not need to write code to implement a priority queue.

An abstract data type is a way of describing data structures that instead of focusing on how they are implemented treats them as types and focuses on the values that can exist of that type and specifically the operations that can be performed on them.

Taking the example of a priority queue, it might be implemented in many ways - with arrays ( in different ways), lists, etc. However, as an abstract data type we would focus on the operations: what are the fundamental things we can do to a priority queue, from which any other activity using it can be built. 

For a priority queue there are three main operations: 

- to add an element to the queue (with a given priority);
- check which element is first; and
- to remove an element from the queue.

What matters is that the highest priority element in the queue is always the next one removed. Other operations are also needed: 

- to create an empty queue and
- to check whether the queue is empty (so we do not try to remove something when nothing is there).
- Similarly it may be useful to have an operation to check if the queue is full.

- With these operations we can implement programs that use the priority queue, irrespective of how the priority queue itself is actually implemented. How that is done is totally hidden when coding a program that uses it.

These operations can each be implemented via methods in Java. For example, having created a new type Priority Queue (as a record) and so some underlying implementation assuming the queue holds strings, and priority is specified as an integer (however priority is implemented internally) we might provide methods with signatures
```java
public static PriorityQueue InsertInPriorityQueue(PriorityQueue p, String toinsert, int priority)
public static String PeekAtPriorityQueue(PriorityQueue p)
public static PriorityQueue RemoveFromPriorityQueue(PriorityQueue p)

public static PriorityQueue CreatePriorityQueue(int max_size)

public static boolean CheckIfEmptyPriorityQueue(PriorityQueue p)
public static boolean CheckNotFullPriorityQueue(PriorityQueue p)
```
The first method will give us a way to add to a priority queue, returning the new updated priority queue; the second allows us to see what is at the front, giving back the string there, and the third removes the highest priority element from the front of the queue returning the new queue.

The final three create a new empty queue (returning the new queue), check if it is empty or check if it is full (each returning a boolean).

Those method specifications are all a programmer coding the rest of the program needs to know. They do not need to know how they are coded or any further details of the underlying data structure. Those methods are also the only parts of the program that actually access the underlying data structure too, however it is implemented. Everything else in the program would call them as needed. 

Doing it this way allows their implementation to change without any effect on the rest of the program.

## Question 5 \[10 marks\]

**Justify** the statement "Abstract data types help make programs easier to write, understand and change." **Illustrate** your answer using the example of a priority queue from the previous question.

### Example answer

Abstract data types (ADT) help make programs easier to write understand and change because of the way they hide unimportant detail from the programmer when they are writing the rest of the program. An ADT if implemented well provides a clean interface to the underlying data structure ensuring it is only used in the way intended. It means that when the programmer is writing a program, once they have implemented the ADT, can forget the detail of how it is implemented. All they need to know are the operations involved: what they do, and their signature (name, return type, number of arguments and argument types). They do  to, not how they do it. They do not even have to have implemented the ADT first. So for example with a priority queue, if they have the specification as given in the headers, they can go ahead coding the program that uses it, knowing that they will not need to change anything later. They know the methods to call to get things done, the arguments they must pass to them and the type of result they will return.

For similar reasons this makes the program easier to understand as someone reading the program only needs to know what the operations (like InsertInPriorityQueue) do to understand code that uses them. They don’t need to work through the detail of how they do it. They do not have to look at the priority queue implementation to understand the rest of the program. Similarly they can look at, and understand the priority queue code without having to read any of the rest of the program.

In terms of changing the program, first making it easier to understand means changes are easier and less likely to involve mistakes anyway, but specifically an abstract data type approach also allows the implementation of that ADT to be changed without any changes at all being needed in the rest of the program. It should just work as the interface provided is unchanged. It also means anyone making such changes only needs to focus on the ADT itself, making sure the operations do the same thing after the change as they did before even though done in a different way.

With a priority queue for example, to change the way it is implemented we only need to change those 6 methods. We can test the new implementation totally separately from the rest of the program. Once they work with the new implementation, then the rest of the code will do exactly as it did before. So for example we could initially implement it as a series of arrays, one per priority level (stored in a record). We might later change this to a single  list where higher priority elements on being added are shuffled to the correct position at the front of the list. This might slow down the operation of adding  to the priority queue, but would use memory more efficiantly perhaps. However in terms of correctness it would do exactly the same thing. This new implementation of the priority queue could be tested using the same test code as originally to ensure the methods work correctly together, and only then swapped into the main program.

## Question 6 \[10 marks\]

**Write** a program to answer queries about the zones of overground stations. It should provide a menu running in a loop that gives options of adding a new station, looking up the zone of a named station, or quitting. The underlying information about stations (pairs of names and zones) should be stored as an array of station records. However it should use a procedural programming abstract data type approach to implement this, ensuring that the underlying representation can easily be changed.

### Example answer

````java
/* ***************************************************
Create a new type (a record) to store pairs of stations and the zones they are in
*/

class Station
{
    String station; // The name of the underground station
    int zone;   // The zone it is in
} // END class Station
````

````java
/* *************************************************** */
/* *************************************************** */

/* ***************************************************
Create a new type (a record) to represent a database of station information
*/

class StationDatabase
{
    Station[] stationarray; // A whole database of info about stations
    int nextfree; // the next unused entry in the database
} // END class StationDatabase	
````

````java
/* ***************************************************
AUTHOR Paul Curzon

Maintain a database of station details allowing 
the zone of different stations to be looked up
*/
import java.util.*;

class StationZones
{

    public static void main(String []p)
    {
        boolean quit = false;
        StationDatabase sdb = createStationDatabase();
       
        while (quit == false)
        {
            String choice = getChoice();
            if (choice.equals("Add"))
                sdb = addEntry(sdb);
            else if (choice.equals("Find"))
                findEntry(sdb);
            else if (choice.equals("Quit"))
                quit = true;
        }
       
        return;
    } // END main
    
    public static String input(String message)
    {
        String ans;
        Scanner scanner = new Scanner(System.in);
      
        System.out.println(message);
        ans = scanner.nextLine();
        return ans;
   } // END input

    public static String getChoice()
    {
        return input("Choose Add, Find or Quit");
    } // END getChoice

    /* *************************************************** */
    /* *********STATION DATABASE METHODS****************** */
    /* *************************************************** */

    public static void findEntry(StationDatabase sdb)
    {
        String key = input("What should I find?");
        int zone = getZoneOfStation (sdb, key);
      
        if (zone == -1) 
            System.out.println("No Idea");
        else
            System.out.println("It is in zone " + zone);
          
        return;
   } // END findEntry

    // Add the details of a new station to the database
    public static StationDatabase addEntry(StationDatabase sdb)
    {
        String name = input("What is the name of the station?");
        int zone = Integer.parseInt(input("What is the zone of the station?"));
        Station s = createStation (name, zone);
     
        addtoStationDatabase(sdb, s);
          
        return sdb;
    } // END addEntry

    // Set the name of a station returning the updated record
    public static StationDatabase createStationDatabase()
    {
        StationDatabase db = new StationDatabase();
        db.stationarray = new Station[10];
        db.nextfree = 0;
        return db;
    } // END createStationDatabase
 

    // Add the given new station to the record. Note will overwrite 
    // last added when full.
    public static StationDatabase addtoStationDatabase(StationDatabase db, Station s)
    {
        Station [] sarray = db.stationarray;
       
        if (db.nextfree < sarray.length)
        {
            db.stationarray[db.nextfree] = s;
            db.nextfree = db.nextfree + 1;
        }
        return db;
   } // END addtoStationDatabase


    // Lookup a given station and return the zone.
    public static int getZoneOfStation(StationDatabase db, String s)
    {
        int zone = searchStation(db.stationarray, db.nextfree, s);
        return zone;
    } // END getZoneOfStation

    // Search an array of Stations and return zone found or -1 if not found
    public static int searchStation(Station [] a, int length, String key)
    {
        for (int i = 0; i < length; i++)
        {
            if (getStation(a[i]).equals(key))
                return getZone(a[i]);
        }
        return -1;
   } // END searchStation


    /* *************************************************** */
    /* *********STATION METHODS*************************** */
    /* *************************************************** */

    
   // Set the name of a station returning the updated record
   public static Station createStation(String name, int zone)
   {
     Station s = new Station();

     s = setStation(s, name);
     s = setZone(s, zone);

     return s;
   } // END createStation

    // Return the station name from a station record
    public static String getStation(Station s)
    {
        return s.station;
    } // END getStation

    // Return the zonefrom a station record
    public static int getZone(Station s)
    {
        return s.zone;
    } // END getZone
 
    // Set the zone of a station returning the updated record
    public static Station setZone(Station s, int zone)
    {
        s.zone = zone;
        return s;
    } // END setZone

    // Set the name of a station returning the updated record
    public static Station setStation(Station s, String stationname)
    {
        s.station = stationname;
        return s;
    } // END setStation
    
} //END class StationZones 
````
