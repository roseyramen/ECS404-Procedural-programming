## Question 1 \[10 marks\]

**Write** a program fragment that uses a while loop rather than a for loop that does exactly the same as the following for loop in all circumstances. **Justify** the equivalence of the two fragments.
```
for(int i = 1; i < 4; i++)
{
    System.out.println(i);
}
```

### Example answer


````java
int i = 1;
while(i < 4)
{
    System.out.println(i);
    i = i+1;
}
````

The for loop first does its initalisation step, before entering the loop executes   
`int i = 1;`  
so this needs to go outside the equivalent while loop. Next the test is done. Both while and for loop have a test that says keep going while the test is `true` so exactly the same test can be used in the while loop  
`while(i < 4)`  
For the body of the while loop we need the instructions that are repeated every time. That is the body of the for loop but also the step that increments the counter. In the for loop `i++` is used this is equivalent to `i = i+1;` so either could be used in the while loop body. On exiting, the for loop then executes whatever comes after the loop, as does a while loop so nothing special needs to be added.

## Question 3 [10 marks]

**Write** a Java program that uses a while loop to simulate a dice game. The person throws a real (6-sided) dice and types in the number they throw, with the program keeping score. The person aims to gain the highest score they can by repeatedly throwing the dice. If they roll a six they get score 0 and the game ends, but if they decide to stop before rolling a six their score is the sum of all their throws.

An example game might be:

`Would you like to throw?` **`y`**  
`What number did you throw?` **`4`**  

`Your total is 4.`

`Would you like to throw?` **`y`**  
`What number did you throw?` **`2`**

`Your total is 6.`

`Would you like to throw?` **`n`**

`You scored 6.`

### Example answer

````java
/* A simple dice game 
   AUTHOR ANON (SID 231234567)
*/

import java.util.*;

class diceGame
{
    public static void main(String [] param) 
    {
        int total = 0;
        
        total = playGame(total);
        printTotal(total);
        
        return;  
    } // END main

    // Print the final total
    //
    public static void printTotal(int total) 
    {
        System.out.println("You scored " + total + ".");
    } // END printTotal

    // Play game repeatedly rolling dice and  return score
    //
    public static int playGame(int total) 
    {
        String ans = input("Would you like to throw (y/n)?");
        
        while(isYes(ans))
        {
            /* roll dice */
            int mythrow = rollDice();
            total = newTotal(total, mythrow);
            
            System.out.println("Your total is " + total + ".");
            
            ans = checkIfContinue(mythrow);
        } // END while loop

        return total;  
    } // END playGame

    // Is it yes (y Y, Yes, YES, yes)
    //
    public static boolean  isYes(String answer)
    {       
        if ((answer.equals("YES")) || 
            (answer.equals("Y")) ||
            (answer.equals("y")) ||
            (answer.equals("yes")) ||
            (answer.equals("Yes"))
           )
        {
           return true;
        }
        else         
        {  
            return false;
        }
    } // END isYes


    
    // Check to see if continue. A 6 automatically means stop so no 
    //
    public static String checkIfContinue(int new_throw)
    {
        String carry_on;
    
        if (new_throw == 6) /* If 6 quit by overwriting ans to no */
        {
            carry_on = "n";
        }
        else // Not 6 so see if they want to continue
        {  
            carry_on = input("Would you like to throw (y/n)?");
        }
        return carry_on;
    } // END checkIfContinue
    
    // Calculate new total 
    //
    public static int newTotal(int total, int new_throw)
    {
        int new_total;
        
        if (new_throw == 6) /* If its a 6 then no score */
        {
            new_total = 0;
        }
        else // Not a 6 so update the total,        
        {  
            new_total = total + new_throw;
        }
        return new_total;
    } // END newTotal
 
    // Roll dice 
    //
    public static int rollDice()
    {
        int my_throw = inputInt("What number did you throw?");

        while ((my_throw < 1) || (my_throw > 6))
        {
             my_throw = inputInt("What number did you throw (it must be 1-6)?");
        }
        return my_throw;
    } // END rollDice
 
    // Input Integer
    //
    public static int inputInt(String message)
    {
        return Integer.parseInt(input(message));
    } // END inputInt

    // Input strings
    //
    public static String input(String message)
    {
        Scanner scanner = new Scanner(System.in);
        String ans;

        System.out.println(message);
        ans = scanner.nextLine();
   
        return ans;
    } // END input

} // END class DiceGame
````
