## Refresher: References

We have seen that simple values (integers, booleans, doubles, etc) are stored directly on the stack but more complex data structures (arrays, Strings, records) are stored as a Reference on the stack pointing to a place on the heap where the actual data is stored. We have also seen how this explains various things that otherwise seem odd to do with assignment and equality, for example.

This way of storing values as references has ramifications for methods. Different behaviour results depending on whether a direct value or a reference is passed to a method.

## Call by value

What is happening in the above is a form of parameter passing called call by value. It is a particular mechanism to pass data to a method.

It is the value stored in a variable that is the actual parameter passed to a method. The value passed is stored in the formal parameter of the method. That formal parameter is a local variable to the method. So in the above, it is the contents of y, the value 550 that is the actual parameter passed. It is stored in local variable x. Changing x has no effect on y.

When a method returns, its stack frame (the area on the stack containing all its variables) is removed. All its local variables disappear along with values stored in them. In the above x and its final value, 551, is lost when the method returns. Only y now exists and as it is a different variable to x, it is unchanged by the method call. 

We could only change y by having the method return a value and store it back in y using an assignment.

Other languages have other mechanisms for passing arguments that work in completely different ways to call by value. Call by value is the simplest and purest in the sense that it is easiest to reason about mathematically so has been adopted widely as a core way values are passed to methods. Other ways have been dropped as they can make code harder to understand.

## Pass by Reference versus Pass by Value

Array parameter passing is still being done as call by value: it is still just the *value* on the stack that is being passed. However, with arrays it is just that the value there, so passed, is a reference. We refer to this as **pass by reference** when a reference is passed using call by value, as opposed to **pass by value** when a primitive value is passed using call by value. In the former the value being passed is a reference. In the latter the  value being passed is not a reference.

SOme languages have an actual built in mechaism for doing **call by reference** where the method can specify that a reference is passed so the original variable changed. Then it is the programmers choice whether a reference is passed or a copy of the array is made. The programmer gets to choose: then  if they choose call by value a copy of the data is passed. Java does not allow this , however.

Here, there are two variables, my_array and a that end up holding copies of the same reference. The two variables are in different stack frames. The variable my_array only exists in the stack frame of the method arrayInc. When that method returns that stack frame is destroyed and the variable lost. However, it was pointing to the place on the heap, and that is not destroyed when the method returns. Values on the heap stay for as long as something somewhere is pointing to them. The original variable, a, on the earlier stack frame is still pointing to that array data. When the method returns therefore, it sees the data changed.

The sequence that happens in the above is ...

Before arrayInc is called:

a ------> 550

When arrayInc starts to run (a new stack from is created for the method containing my_array and it is initialised with the reference in a).

a ------> 550 <------- my_array


After my_array\[0\] is changed

a ------> 551 <------- my_array

After arrayInc returns (its stack frame is removed so my_array disappears).
a ------> 551

## Key points about arrays and methods

The important thing to notice in the above examples is that when using pass by reference, the method does NOT return a value but still changes the values in (one or more of) the array argument(s) passed.

This is all because an array is stored as a reference.

We thus have a new way that a method can communicate results back to the calling method. Both of the  methods below, for example, work as a way to initialise an array, as the test calls show.

````java
// Initialise an array to all Xs relying on pass by reference
//
public static void initArray1 (String[] array_to_init)
{    
    for (int i = 0; i < array_to_init.length; i++)
    {
        array_to_init[i] = "X";
    }
    return;
} // END initArray1
````

## Different styles of programming


Some communities / languages (so style guides) advocate the first style which is functional in nature. Others advocate the second which is procedural. The functional style is mathematically pure and makes code easier to reason about as it avoids methods having side effects, changing things invisibly in method calls. The procedural style in some ways is more direct (and involves less typing for what that is worth!) It is also the style that forms the basis of object-oriented programming which is a different style again (actually a whole different paradigm). In some situations it can avoid the copying of large amounts of data.

You certainly should NOT mix the styles in a program.

You should be able to program either way.

### Solution to Exercise 3<a id="Solution8-3-3"></a>


This prints
```
Before I call intArray my_array[0] is 550
The value in a[0] is 551
After I call intArray my_array[0] is 551
```
The method has somehow changed the values not just in the array that is its local variable but also the array that was passed as an argument. That variable does not exist inside the arrayInc method. So how is it changed by the method?

The array value as a whole is not passed to the method. Instead a reference to the place on the heap where the array data can be found is passed. That is because the actual array value is just stored as a reference on the stack. The value in that variable location is being passed just as with any other variable. What is passed is always a copy of the bit pattern found in that variable's location on the stack. It is just the fact that it is a reference that changes things.

The reference that is passed is stored in the local variable a on the stack. No copy of the actual array data has been made, only a copy of that reference. Now though both variable a and variable my_array are pointing to the same place on the heap as both hold the same reference. 

Therefore, when the method refers to `a[0]` this is telling it to go to variable a, where it will find a reference (a pointer) and to follow where that reference is pointing. Once there it goes to the first entry of the array value found there. That has directed it to the place that my_array is also pointing to - the original place the array data was stored (not a copy of it). Once there it changes it. When it goes there to get a copy of the value (on the right hand side of the assignment), nothing is different in its behaviour as right-hand side uses do not change anything. However, the left-hand side of the assignment accesses that location and stores a new value back there (again getting there by following the reference).

It does not appear to behave like intInc because with intInc a copy of the actual data was passed (the value 550), rather than as here a copy of the reference to the actual data.

### Solution to Exercise 8<a id="Solution8-3-8"></a>



Both methods successfully initialise a String array passed as an argument, filling all entries with the String "X". How they do this differs, however, as the first method, initArray1, uses the fact that the array is passed by reference to initialise the array argument. The second, on the other hand, works as though pass by value is happening. In both cases it is a reference that is passed as that is how arrays are stored on the stack (as a reference). This means that in reality, the call to initArray1 could ignore the result and the array would still have changed.

The key difference is therefore that the first method is written to support a functional style of method use, whereas the second is written in a pure procedural style, relying on side effects to change the array. The second does not actually return any value, whereas the first returns a String array value ie a reference to the resulting initialised real array data stored on the heap. This array value returned is then used to overwrite the original one in the call by the assignment:
```java
a2 = initArray(a2);
```
The method thus returns the reference of the array value and that reference is placed back in a2. In fact, it is exactly the same array reference as was already there, because it is the same reference that was passed as the argument.

An advantage of the functional style is that what it does (returning an updated array) is made clear in the signature (ie header line) of the method. It is also clear in the call that it updates the array, a2. By contrast, in the procedural style its side effect is only clear on reading, and understanding, the whole method. It is not clear at all from the call that anything changes. The functional style of the second method thus arguably makes it easier for a reader to understand what the method does.

In summary the two methods both do the same thing ultimately, but just use a different style of programming.

### Solution to Exercise 9<a id="Solution8-3-9"></a>

It would be even better if the functional style method to initialise an array didn't do the side effect of changing the array on the heap at all. There is no need in this case to actually pass the array as an argument, in fact, as the values in it are not used. Instead, a new array value on the heap could be built such as in the following:

```java
public static String[] initArray3 (int length)
{    
    String[] temp_array = new String[length];

    for (int i = 0; i < length; i++)
    {
        temp_array[i] = "X";
    }
    return temp_array;
} // END initArray3
```
This is better as it does not change anything not local to the method as a side effect and so is purely functional. There is no possibility of it being used in a away so that it changes an array unintentionally due to the side effect. Its job is to create a particular array value (of a given length), here containing all Xs. That is all this method does. What is done with that array value is completely down to the call and what is done with the result returned. This makes it mathematically pure so much easier to reason about.

This is not a general solution that can be used in all situations, however. It does not work if the values in some array passed are to be used. The equivalent then would be to be pass the array argument but copy the values from it into a new array, so that no change was made in the method to the original actually passed as an argument. This could waste both a lot of time and of memory though in doing the copying.

````java
// Initialise an array value to all Xs using a pure functional style.
// Construct a new array value of the given length to do this.
//
public static String[] initArray3 (int length)
{    
    String[] temp_array = new String[length];

    for (int i = 0; i < length; i++)
    {
        temp_array[i] = "X";
    }
    return temp_array;
} // END initArray3
````
