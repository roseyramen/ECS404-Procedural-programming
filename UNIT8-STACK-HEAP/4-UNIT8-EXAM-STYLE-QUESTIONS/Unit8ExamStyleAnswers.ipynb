## Question 1 \[6 marks\]

**Explain** what is meant by  
i) the stack  
ii) the heap  
with respect to the Java runtime environment.

### Example answer

i) The stack is an area of memory used by the Java runtime system. When variables are declaredthey are allocated storage space (essentially a word) on the stack. It works like a stack in that thenew space allocated grows upwards in an organised way - the next variable will be allocated thememory location directly after the last one. Each time a new method is called a new stack frame isstarted and when the method returns the stack returns to that state in that all the variables declaredare unallocated. Thus it acts as a stack at the level of method calls and stack frames. This mirrors what is needed for methods being called and returned to wherever they were called from. 

The values of simple typed variables like ints and chars are stored on the stack.
eg  
```java
int a = 3;`   
char b = 'a';`
```
Would both store a value (`3` and `'a'`) on the stack and in consecutive positions.

More complex typed variables (actually Objects in Java), including strings, records and arrays have a reference stored on the stack with the data on the heap.  A reference is just a pointer to the place in memory where the data is actually stored. References also take up a word of memory, so in essence everything stored on the stack is the same size and the stack builds up. This means that variables are stored at a predictable position from the start of the stack frame. These more complex types are called reference types as a result of the way they are stored as a reference on the stack.ii) The heap is a less structured storage area separate to the stack and allocated by the run time system.Essentially it holds data of variable sizes (ie data that doesnt fit into a word). The actual data of reference typed variables, so arrays, records, strings (and objects more generally) are stored on the heap (though when declared as stored in a variable with references to them onthe stack so they can be found). 
For example
```java
int[]a = {1,2,3};
```
On the stack this would allocate a space for a reference in the next free position. It would point to the
place on the heap where the values 1, 2 and 3 could be found ie where the array data actually is.

The heap can also store references. For example, an array of records would be stored as a reference in the stack pointing to an array on the heap but that array on the heap would contain references to other places on the heap where all the records would be found.

Places on the heap are allocated by the system finding memory that is free and that islarge enough for the current purpose. This is done by the new keyword used when the reference type is initialised. Thus, newly allocated space could be any where in the heap. InJava a garbage collector deallocates storage on the heap, freeing it up when nothing is pointing to it anymore so that is is inaccessable.

## Question 2 \[10 marks\]

**Compare and contrast** the following, illustrating your answer with examples:  
i) method call with an array parameter   
ii) method call with an integer parameter

### Example answer

The most obvious difference between method calls with array and integer parameters is just in thetype of the argument. In the first case a whole array is passed and in the second a single integer.However, because integers are stored directly (on the stack) in integer variables, whereas arrays arestored in two parts (a reference on the stack and the data on the heap) they behave differently.    

Suppose we have a method: 
```javapublic static void add2(int a){    a = a + 2;}
```Then a call
```javaint b = 5;add2(b);
```passes the contents (ie value) of `b` (ie `5`) to the method and this is stored in its formal parameter `a`. `a`and `b` are different variables stored in different places on the stack. That means when the methodadds `2` to `a`, it changes `a` but has no effect on `b` at all. When the method returns, `a` disappears whenits stack frame is removed and `b` still holds the value `5`.
Contrast this with a similar method taking an array parameter:  
```javapublic static void add2(int[] a){    a[0] = a[0] + 2;}
```Then a call  
```javaint[] b = {5};add2(b);
```has a different effect. Again the value stored in `b` is passed to the method and stored in `a`. Howeverwhat is stored in `b` is just a reference to a location on the heap (where the value `5` is stored). Whenthe method does the assignment it changes `a[0]`: that is it goes to `a`, follows the reference to the heapand changes the value there. The `5` changes to `7`. Now however `b` is pointing to the same place onthe heap as `a` was - it just has a copy of the reference. As before when the method returns `a` (on thestack) is discarded, but `b` is pointing to the same changed place. The value stored in array `b` haschanged.
The effect of this is that when array variables are passed as arguments, the passed array can bechanged by the method. By contrast, when integer variables are passed the variable will not bechanged. For an integer method to return a value it must explicitly use return whereas an array thatis passed does not need to be explicitly returned

Despite that difference in effect in both cases the value directly in the variable is passed in exactly the same way, and copied into the formal parameter in both cases. It is just that in the int case that is the actual integer value stored and passed, whereas with an array the value directly in the variable (on the stack) is a reference value to the array data not the actual array data itself so it is a reference value stored and passed..

## Question 3 [10 marks]

**Write** a program that allows a user to input a series of names and telephone numbers. Once the names and numbers have been input the program should allow the user to repeatedly search for numbers of people with a name given by the user. The program should use a separate method for inputting the details and for doing the search, both called from the main method. The array(s) used to store the data should be declared in the main method and passed to the other methods as arguments. Your methods should not return array results however but make use of pass by reference parameter passing.


### Feedback on the kind of answer expected

*The following table gives an indicative idea of what marks an answer deserves:*

**A+: 9-10 marks:**

- Very good decomposition into methods based on subtasks
- Outstanding style
- Input validation
- ...or other aspects that add to elegance.
- Logically correct program. It does the right thing.

**A: 7-8 marks:**

- Logically correct program. It does the right thing.
- Greater marks for doing it elegantly...comments, efficient algorithm, methods etc.

**B: 6 marks:**

- Does something, but not quite the right thing.
- Contains at least one logical error or multiple syntax errors, but basically shown can solve problem.

**C: 5 marks:**

- Does something, but not quite the right thing.
- Contains more logical errors, but basically shown can solve problem.


**D: 4 marks:**

- More major errors but evidence that can write a program using appropriate
- constructs if put in front of a compiler - though it may take time.

**F/G: 0-3 marks:**

- Fragmentary code answer - eg rote-learnt bits of code.
- No clear evidence can write a program to solve a problem.

### Example answer

````java
/* ************************************************/
/* Get phone numbers and allow them to be searched */
/* ************************************************/
import java.util.*;

class PhoneSearch
{
    public static void main(String [] param)
    {
        final int MAX_NAMES = 100;
        int NAMES = 0;
        String[] names = new String[MAX_NAMES];
        String[] numbers = new String[MAX_NAMES];
        
        /* First fill the database with numbers
        then switch to routine allowing searches
        */
        NAMES = storePhoneNumbers(names, numbers, MAX_NAMES);
        searchPhoneNumbers(names, numbers, NAMES);
        
        return;
    } // END main
    
    // Store a series of phone numbers while the person wants to continue
    //
    public static int storePhoneNumbers(String[] names, String[] numbers, int SIZE)
    {
        String keepgoing = "y";
        int count = 0;
        
        while(keepgoing.equals("y") & count < SIZE)
        {
            names[count] = input("Next name");
            numbers[count] = input("Next phone number");
            count = count + 1;
            keepgoing = input("Continue (y/n)");
        }
        return count;
    } // END storePhoneNumbers
    
    // Search for a series of phone numbers given names
    // while the person wants to continue
    //
    public static void searchPhoneNumbers(String[] names, String[] numbers, int SIZE)
    {
        String keepgoing = "y";
        int count = 0;
        
        while(keepgoing.equals("y"))
        {
            String name = input("Name to search for");
            int found = search (name, names, SIZE);
            if (found == -1)
                System.out.println("Not known");
            else
                System.out.println("The number is " + numbers[found]);
            keepgoing = input("Continue (y/n)");
        }
        return;
    } // END searchPhoneNumbers
    
    // Search a String array for a given key up to position SIZE.
    //
    public static int search(String target, String[] search_array, int SIZE)
    {
        for (int i = 0; i<SIZE; i++)
        {
            if (search_array[i].equals(target))
                return i;
        }
        return -1;
    } // END search

    // Ask for a string with given message
    // Return the string typed in by the user
    //
    public static String input(String message)
    {
        Scanner scanner = new Scanner(System.in);
        String answer;
        
        System.out.println(message);
        answer = scanner.nextLine();
        
        return answer;
    } // END input

} // END class PhoneSearch
````
