## Question 1 \[10 marks\]

**Explain** what is meant by  
i) the stack  
ii) the heap  
with respect to the Java runtime environment.

### Example answer

i)  
By persistent storage we mean storage of data that survives the end of execution of the program. Data stored in a variable in a program survives only while the program is executing that section of the program after which it is inaccessible. When a program terminates all the data stored throughout its execution is inaccessible. Files are used for persistent storage. This is important so that data can be shared between programs and saved for further sessions. For example a word processor would be little more than a glorified typewriter without the ability to save and reload document files. Saving the document typed as a file means it is preserved even when the word processor program is shut down so it is persistent in that sense of existing for longer then the program run. This means it can be reloaded into the word processor at a later time when it is run again.

ii)   
A text file is a data structure that stores sequences of characters. It is a persistent data structure (as above) in that it persists the termination of the program that creates it. It is stored in the file system rather than in storage local to the runtime environment of the program. The characters in a text file can be accessed in order. That is in the order they are stored in the text file. The characters are placed in a text file, and recovered from them a single character at a time. This can include control characters representing things like new lines.

iii)  
ASCII is a particular representation for storing characters. All data in a digital computer is stored as bit patterns. Depending on the type of data the bit pattern is interpreted in a particular way. ASCII is one way of interpreting bit patterns as characters. Each letter of the alphabet for example has a particular bit pattern (one of 128) in the ASCII encoding. As bit patterns also correspond to numbers it is equivalent to saying that each ASCII character corresponds to a number between 0 and 127. The ASCII for a space character is the bit pattern for number 32, for example, so we can think of 32 meaning space in ASCII> ASCII is being superseded by UNICODE which consists of vastly more bit patterns and so can encode more than just the common English characters.

## Question 2 \[10 marks\]

**Compare and contrast** file input with keyboard input using scanners.

### Example answer

Both are a way of a data obtaining a stream of characters. In the case of file input the stream of characters exist in advance, having been creating previously by some other program. In the case of keyboard input they are created dynamically on demand. For the latter, the user of the program supplies the data on request. This means that the program needs to give instructions on the format and kind of data needed at each point. For file input the data already exists. Either it will be stored in a specific format for the program (such as comma separated files for spreadsheets), or the program must be able to handle whatever is there. 

Both in essence process a stream of characters that can only be processed sequentially. In both cases streams need to be set up in advance to link a program variable to the source of input, the data is processed and then the channel is closed.

With keyboard input the stream is linked to a specific place that is set up by the library - linking the keyboard into an internal stream - System.in. Buffering of characters is dealt with with the system waiting for the enter key.

```java
Scanner scanner = new Scanner(System.in);
```

Once the scanner variable is initialised it can be used to, for example, read whole lines of text using eg

```java
String answer = scanner.nextLine();
```

With file input you make the link from the file yourself, first creating a FileReader - an object that   gets characters from the file, and linking it to a new BufferedReader that buffers characters and so is able to, for example, read whole lines at a time.

```java
BufferedReader inputStream = new BufferedReader(new FileReader(“hello.txt”));
```

The general idea is the same but a difference is that with keyboard input the location is thus set up already, but with a file it is set up using a FileReader based on a given file name.

In both cases the stream of characters created is represented as a variable, though the specific type is different in the two cases (Scanner versus BufferedReader)

Once this is set up reading characters or Strings is similar and done with equivalent methods linked using the dot notation to the variable holding the stream.

```java
String nextword = inputStream.readLine();
```
versus
```java
String nextword = scanner.nextLine();
```
Note the structure of these statements is identical.

In these examples, scanner and inputStream play very similar roles. Once set up the process of getting a character from the stream and storing it in a variable using an assignment is very similar.
Whereas a file must be explicitly closed, the keyboard input stream does not need to be (and normally is not)
.

## Question 3 [10 marks]

**Write** a program that allows a user to input a series of names and telephone numbers. Once the names and numbers have been input the program should allow the user to repeatedly search for numbers of people with a name given by the user. The program should use a separate method for inputting the details and for doing the search, both called from the main method. The array(s) used to store the data should be declared in the main method and passed to the other methods as arguments. Your methods should not return array results however but make use of pass by reference parameter passing.


### Feedback on the kind of answer expected

*The following table gives an indicative idea of what marks an answer deserves:*

**A: 7-10 marks:**

- Logically correct program. It does the right thing.
- Greater marks for doing it elegantly...outstanding style,
- efficient algorithm, methods etc.

**B: 6 marks:**

- Does something, but not quite the right thing.
- Contains at least one logical error or multiple syntax errors, or some small poor style issue. but basically shown can solve problem.

**C: 5 marks:**

- Does something, but not quite the right thing.
- Contains more logical errors, but basically shown can solve problem.


**D: 4 marks:**

- More major errors but evidence that can write a program using appropriate
- constructs if put in front of a compiler - though it may take time.

**F/G: 0-3 marks:**

- Fragmentary code answer - eg rote-learnt bits of code.
- No clear evidence can write a program to solve a problem.

### Example answer

````java
/* *********************************
   AUTHOR 123456789 
   DATE 16/11/23
   check how often each word from a list provided first appears in a named file

*/       

import java.io.*;
import java.util.*;

class CountWords
{
    public static void main(String[] params) throws IOException 
    {
        final int WORDCOUNT = 10;
        final String FILENAME = "wombat.txt";
       
        int[] counts = new int[WORDCOUNT];
        String [] searchwords = new String[WORDCOUNT];

        zeroArray(counts);
        getWords(searchwords);

        checkForSearchWords(FILENAME, searchwords, counts);
        printCounts(searchwords, counts);

        return;
    } // END main

    /* *********************************
       check how many times each word in a given array appears in a named file
    */       
    public static void checkForSearchWords(String filename, String[] words, int counts[]) throws IOException
    {
        BufferedReader inStream = new BufferedReader(new FileReader(filename));
        String nextword = inStream.readLine();

        // Repeatedly check the current word - is it in the array of words
        while (nextword != null)
        {
            for (int i = 0; i<counts.length; i++)
            {
                if (nextword.equals(words[i]))
                    counts[i] = counts[i] + 1;
            }
            nextword = inStream.readLine();
        }
        inStream.close();
        return;
    } // END checkForSearchWords
    
    /* *********************************
        print times each word appears taking words and counts from two arrays passed
    */   
    public static void printCounts(String[] words, int counts[])  
    {
        String result = "";
        for(int i = 0; i<words.length; i++)
        {
            result = result + "There are "+counts[i] + " " + words[i] + "'s.\n";
        }
        System.out.println(result);
        return;
    } // END printCounts
    
    /* *********************************
       Ask for a string with given message Return the string typed in by the user
    */ 
    public static String inputString(String message)
    {
        Scanner scanner = new Scanner(System.in);
        String answer;

        System.out.println(message);
        answer = scanner.nextLine();
        return answer;
    } // END inputString

    /* *********************************
       Get words to search for
    */   
    public static void getWords(String[] a)  
    {
        for(int i = 0; i<a.length; i++)
        {
            a[i] = inputString("Search word " + i);
        }
        return;
    } // END getWords

    /* *********************************
       set all entries in an int array to 0
    */   
    public static void zeroArray(int[] a)  
    {
        for(int i = 0; i<a.length; i++)
        {
            a[i] = 0;
        }
        return;
    } // END zeroArray
    
} // END class CountWords

````
