## Introduction
We have seen how **while** loops can control repetition in a way that the number of times the body of the loop is executed (the number of **iterations**) is unpredictable. While loops should be used whenever the number of iterations is unknown. A **for** loop, by contrast, is best used when the number of iterations of the loop is known when it starts.

In particular, we have seen how we can write a while loop that keeps asking the user if they want to continue, and so keeps going until the user indicates that they want to stop.

We have also seen how a while loop can actually do anything a for loop can do. In particular, you can write a counter- controlled loop using a while loop. However, counter-controlled loops are better written using a for loop as it immediately indicates to a reader that it *is* a counter-controlled loop so is easier to understand. Also, by pulling the three key parts of a counter-controlled loop into one line (initialising the counter, testing the counter and changing the counter), means they are less likely to be forgotten.

Now, we look at how we can have more complex tests of when the for loop ends.

## Logical And (&&) ##

We can use a similar approach using a logical AND connective if we want to keep going for different reasons (or put another way, there are several reasons we might stop). One common situation is where we want a loop to be controlled both by some other stopping condition AND by a counter. We then might use the logical AND connective (&&).

Also the Logical Not connective (!) is often useful, when you need a loop to keep going when some condition is not true.

This gives us a way to do things like combine a counter-controlled loop with some other reason for stopping the count early as we see below. This is NOT done using two loops (as novices often assume), but with a single loop with a more complex condition having more than one reason to stop.

### Solution to Exercise 8<a id="Solution8"></a>

The test is

 isCryForHelp(answer) && (count < 3) 
 
 This means the loop will keep going as long as BOTH
 
 - a cry for help is given (eg HELP is typed when asked) AND
 - the count is less than 3. 
 
The loop will stop therefore if one or the other of these conditions is false. It will stop if the answer was not a cry for help. However, it will also stop if the count becomes greater than or equal to 3.

It will therefore exit the loop and print the count once it gets to 3 or if the person types something other than a cry for help earlier. (So 3 is the highest the count will ever get to and the loop will never execute more than 3 times).

The if statement is needed because before we ask again for the next input from the user we need to be sure we haven't reached the count already - as we only want to ask again if we aren't at the count and so need to stop whatever. However, the while loop test will not actually test the count until we finish the body and jump to the top. So we check whether we need to ask again before we do ask. Notice it is only a single branch if statement as we do nothing extra if the count is reached. Then we just want to continue to the loop test and be released from the loop.

### Solution to Exercise 12<a id="Solution12"></a>

A While loop test can be made more complex by using logical connectives. 

Logical or (||) allows you to combine two different conditions so that the loop keeps going if any of them are true. This is useful for example, when there are various different conditions, any of which are a reason to continue in the loop. As we saw various related input possibilities that would mean "keep going" are one example.

Logical and (&&) is useful when there are a series of conditions, all of which must hold to continue (so the negation of each is a stopping condition). This is useful for example when combining a counter controlled loop with some other earlier stopping condition. 

With multiple stopping conditions care has to be taken that unnecessary computation isn't done after one has become true. This can be an issue because the loop exit condition is only tested when we get to the end of the body and start a new iteration.

It is often good to embed test conditions in methods with names that indicate what the test means (such as hasLost(...)). This, as with any well defined decomposition, makes the code more readable and easier to understand. That method can also then be tested separately.
