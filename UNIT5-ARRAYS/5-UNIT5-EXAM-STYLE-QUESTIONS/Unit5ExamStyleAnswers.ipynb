## Question 1 \[6 marks\]

The following code fragments contain bugs. 
**Identify and correct** the bugs, **justifying** your answers.

i) 
```java
// increment position 5 of array
//
a(5) = a(5) + 1;
```

ii)
```java
// print all elements of an array of size 10
//
for(int i=0; i <= 10; i++)
{
    System.out.println(a[i]);
}
```

iii)
```java
// declare an integer array with 4 elements
//
int [] a = new int[]; 
```

iv)
```java
// Set all elements of an array of size 10 to 300
//
for(int i=1; i < 10; i++)
{
    a[i] = 300;
}
```

v)
```java
// Print the elements of an array of size 10 in reverse order
//
for(int j=0; j < 10; i--)
{
    System.out.println(a[i]);
}
```

### Example answers
i)

Square brackets should be used round the index of an array not parentheses:
```java
a[5] = a[5] + 1;
```

ii)  

This overshoots the end of the array. The 10th element is in position 9. Use `<` rather than `<=` or just `9` instead of `10` in the test:
```java
for(int i=0; i < 10; i++)
{
    System.out.println(a[i]);
}
```

iii)  

The size is missing where the array is created:
```java
int a[] = new int[4];
```

iv)

This loop misses the first element of the array. The first element of an array is not position 1 but position 0. Therefore the loop counter needs to be initialised to 0 not to 1. 

```java
for(int i=0; i < 10; i++)
{
    a[i] = 300;
}
```

v) 

Here j-- is correctly used to decrement the loop counter but the counter starts at 0 and the end test is then 10 (indicating the size of the array). However the count will not get there if counting down from 0. It needs instead to start at 9 (as the last element of the array of size 10 is position 9, so it can count down to 0 with the test being about j being greater than or equal to 0 to stop (so being less than 0 to finish).
```java
for(int j=9; j >= 0; i--)
{
    System.out.println(a[i]);
}
```

## Question 3 [10 marks]


**Write** a Java program that reads 20 positive integers from the keyboard into an array. If an invalid number is read in then the program should store -1 in that position in the array to show this. It should then search the array for a positive integer provided by the user, printing the position in the array of *all* places where that positive integer was found. Your program should use for loops and if statements as well as arrays.



### Feedback on the kind of answer expected

*The following table gives an indicative idea of what marks an answer deserves:*

**A: 7-10 marks:**

- Logically correct program. It does the right thing.
- Greater marks for doing it elegantly...comments, efficient algorithm, use of methods, good decomposition, etc.

**B: 6 marks:**

- Does something, but not quite the right thing.
- Contains at least one logical error or multiple syntax errors, but basically shown can solve problem.
- very good style

**C: 5 marks:**

- Does something, but not quite the right thing.
- Contains more logical errors, but basically shown can solve problem.
- good basic style

**D: 4 marks:**

- More major errors but evidence that can write a program using appropriate
- constructs if put in front of a compiler - though it may take time.
- may have style problems but some indication of good style

**F/G: 0-3 marks:**

- Fragmentary code answer - eg rote-learnt bits of code.
- No clear evidence can write a program to solve a problem.
- poor style
- does not use the required constructs

### Example answer

````java
/* ****************************************
   AUTHOR ANON (123456789)
   
   Read in 20 positive integers from the keyboard then search the array, 
   printing the position in the array of all places where the 
   positive integer given was found 
*/

import java.util.*;
 
class findints
{
  public static void main (String [] param) 
  {
      final int INVALID_P_INT = -1;
      final int SIZE = 20;
      int [] a = fillArray(SIZE, INVALID_P_INT);
      
      int key = inputPositiveInt("What should I search for?", INVALID_P_INT);
      
      if (key == INVALID_P_INT)
          System.out.println("That is not a valid positive integer to search for");
      else
          find(key, a);
      
      return;
        
  } // END main

  // fill an array with positive ints,storing the given invalid number if not given a +ve int
  //
  public static int[] fillArray (int SIZE, int INVALID_P_INT)
  {
    int [] a = new int[SIZE];

    // Load values into array, stores -1 (INVALID_P_INT) if an invalid number is input
    //
    for (int i=0; i<SIZE; i++)
    {
      a[i] = inputPositiveInt("Give me a value to store", INVALID_P_INT);
    }
      
    return a;
  }

  // search array for all occurrences, printing as go
  //
  public static void find (int key, int [] a)
  {
    boolean found = false; //not found it yet!
    for (int i=0; i < a.length; i++)
    {
      if (a[i] == key) 
      {
           System.out.println("FOUND in position " + i);
           found = true;
      }
    }
    
    if (! found) System.out.println("It was not there!");
    return;
  }

  // Input strings
  //
  public static String input(String message)
  {
       Scanner scanner = new Scanner(System.in);
       String ans;

       System.out.println(message);
       ans = scanner.nextLine();
   
       return ans;
  } // END input

  // Check if a character is a digit
  //
  public static boolean isDigit(char c)
  {
      if(c=='0' || c=='1' || c=='2' || c=='3' || c=='4' || c=='5' || c=='6' || c=='7' || c=='8' || c=='9')
          return true;
      else
          return false;

  } // END isDigit

  // Check if a String is a positive integer
  //
  public static boolean isPositiveInt(String invalue)
  {
      for(int i = 0; i < invalue.length(); i++)
      {
          if (! isDigit(invalue.charAt(i)))
              return false;
      }
       return true;
  } // END inputInt
     
  // Input a positive integer, return the given sentinal if not positive int
  //
  public static int inputPositiveInt(String message, int INVALID_P_INT)
  {
      String str_value = input(message);
      
      if (isPositiveInt(str_value)) 
          return Integer.parseInt(str_value);
      else
          return INVALID_P_INT;
  } // END inputInt

} // END class findints
````
