### stringInput

#### What it does
This method prints a message and returns the user's input to the caller

#### Implementation (how it works)
A scanner variable is created and its initialised to a scanner object, we then take in a message parameter and then print that message out. What the user inputs next is copied to a string varaible and then returned to the caller.

````java
    public static String stringInput(String message)
        {
            Scanner scanner = new Scanner(System.in);
            System.out.println(message);

            String userInput = scanner.nextLine();

            return userInput;
        }//end of stringInput
````

### bookPrice()

#### What it does
This method asks the user a question about the number of books in the shops and assigns a book price based on the answer
#### Implementation (how it works)
This method works by setting a range of final variables to please proffessor. We then ask the user about the number of books inside of shop. We then compare the answer to a range of bounds and return a price, if the number of books arent in any of the bounds and an invalid price is sent up which will cause the main program to stop pre-emptivley.

````java
    public static int bookPrice()
    {
        int input = Integer.parseInt(stringInput("How many copies does the shop hold in total?"));

        final int surgePrice = 4;
        final int normalPrice = 3;
        final int discountedPrice = 2;
        final int invalidPrice = -1;
        final int surgeUpperBound = 3;
        final int surgeLowerBound = 0;
        final int normalLowerBound = 3;
        final int normalUpperBound = 4;
        final int discountLowerbound = 4;

            if(input < surgeUpperBound && input > surgeLowerBound)
            {
                return surgePrice;
            }
            else if (input == normalLowerBound|| input == normalUpperBound)
            {
            
                return normalPrice;
            }
            else if (input > discountLowerbound) 
            {
                return discountedPrice;
            }
            else
            {
                System.out.println("That is not a legal number of books. Please try again.");
                return invalidPrice;
            }
    }
    
````
