### growthCalc(int waterCount, int sunlightHours)

#### What it does
    
Takes in the watercount and hours of sunlight parameters and carries out the expected growth calculation and returns it to the caller


#### Implementation (how it works)
Takesin waterCount and sunlightHours in as integer parameters and then calculates the expected growth my mulipling them against constant varaibles and then adding the results togethet, we then return the integer.

````java
    public static double growthCalc(int waterCount, int sunlightHours)
    {    
        final double waterGrowthConstant = 1.2;
        final double sunlightGrowthConstant = 0.5;
        double growth = (waterGrowthConstant * waterCount) + (sunlightGrowthConstant * sunlightHours);

        return growth;
    }
````

##### percentageIncrease(double growth, double startHeight)

#### what it does

    takes in growth and starting height to calculate the percentage change in height

#### how it does it

takes in double parameters in the form of growth and height and then carries out the calculation by dividing growth by startheight and then multpling resut by a hundred and then rounding up. to meet the format requirements we then turn the double into an integer we then change the percentage.

````java
public static int percentageIncrease(double growth, double startHeight)
    {
        double difference = growth / startHeight;
        double percentage = difference * 100;
        percentage = Math.ceil(percentage);
        int integerPercentage = (int)percentage;
        
        return integerPercentage;
    }
````
